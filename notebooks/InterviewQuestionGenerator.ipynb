```python
# STEP 4.1 — Extract Raw Skills

skills_section = sections["Skills"]

raw_skill_entries = []

for line in skills_section:
    # Remove the bullet symbol
    line = line.replace("•", "").strip()

    # Find the text after the category colon
    if ":" in line:
        skill_text = line.split(":", 1)[1].strip()

        # Split individual skills by comma
        skills = skill_text.split(",")

        for skill in skills:
            skill = skill.strip()

            if skill:
                raw_skill_entries.append(skill)

print("Raw skills extracted:\n")

for skill in raw_skill_entries:
    print("-", skill)

print("\nTotal raw skill entries:", len(raw_skill_entries))
```

```python
# STEP 4.2 — Normalize Skills

def normalize_skill(skill):
    """
    Convert different representations of the same skill
    into one standard name.
    """

    skill = skill.strip()

    # Remove version information
    skill = re.sub(r"\s*\([^)]*\)", "", skill)

    # Convert common variations to standard names
    aliases = {
        "react.js": "React",
        "reactjs": "React",
        "nodejs": "Node.js",
        "node.js": "Node.js",
        "html5": "HTML",
        "css3": "CSS"
    }

    normalized = skill.lower()

    if normalized in aliases:
        return aliases[normalized]

    # Preserve normal capitalization for known skills
    known_skills = {
        "html": "HTML",
        "css": "CSS",
        "javascript": "JavaScript",
        "next.js": "Next.js",
        "tailwind css": "Tailwind CSS",
        "express.js": "Express.js",
        "mongodb": "MongoDB",
        "mysql": "MySQL",
        "sql": "SQL",
        "c++": "C++",
        "git": "Git",
        "github": "GitHub",
        "gitlab": "GitLab",
        "vs code": "VS Code",
        "figma": "Figma",
        "postman": "Postman",
        "agile development": "Agile Development"
    }

    return known_skills.get(normalized, skill)


normalized_skills = []

for skill in raw_skill_entries:
    normalized = normalize_skill(skill)

    if normalized not in normalized_skills:
        normalized_skills.append(normalized)


print("Normalized skills:\n")

for skill in normalized_skills:
    print("-", skill)

print("\nTotal normalized skills:", len(normalized_skills))
```

```python
# STEP 6.3 — Extract Experience Duration

from datetime import datetime
import re


def calculate_months_from_dates(date_text):
    """
    Calculate approximate duration in months when a clear
    start date and end date/present date are available.
    """

    date_text = date_text.strip()

    # Pattern: Month Year – Present
    present_pattern = re.search(
        r"(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Sept|Oct|Nov|Dec)[a-z]*\s+(\d{4})\s*[–-]\s*Present",
        date_text,
        re.IGNORECASE
    )

    if present_pattern:
        month_text = present_pattern.group(1).lower()
        year = int(present_pattern.group(2))

        month_map = {
            "jan": 1,
            "feb": 2,
            "mar": 3,
            "apr": 4,
            "may": 5,
            "jun": 6,
            "jul": 7,
            "aug": 8,
            "sep": 9,
            "sept": 9,
            "oct": 10,
            "nov": 11,
            "dec": 12
        }

        start_month = month_map[month_text[:4] if month_text.startswith("sept") else month_text[:3]]

        start_date = datetime(year, start_month, 1)
        current_date = datetime.now()

        months = (
            (current_date.year - start_date.year) * 12
            + (current_date.month - start_date.month)
        )

        return max(months, 0)

    return None


for entry in experience_entries:

    date_text = entry["date"]
    description = entry["description"]

    duration_months = None

    # Case 1: Explicit 8-week duration
    week_match = re.search(
        r"(\d+)\s*-?\s*week",
        description,
        re.IGNORECASE
    )

    if week_match:
        weeks = int(week_match.group(1))
        duration_months = round(weeks / 4.345, 1)

    # Case 2: Date range ending in Present
    elif "present" in date_text.lower():
        duration_months = calculate_months_from_dates(date_text)

    entry["duration_months"] = duration_months


print("Experience durations:\n")

for i, entry in enumerate(experience_entries, start=1):

    print("=" * 60)
    print(f"EXPERIENCE {i}")
    print("=" * 60)

    print("Organization :", entry["organization"])
    print("Role         :", entry["role"])
    print("Date         :", entry["date"])
    print("Type         :", entry["type"])
    
    print("Duration     :", entry["duration_months"], "months")
```

```python
# STEP 7.3 — Normalize Project Technologies

def normalize_project_technology(technology):
    """
    Normalize project technology names while preserving
    important distinctions between different technologies.
    """

    technology = technology.strip()

    # Remove parenthetical details such as "(Expo)"
    technology = re.sub(r"\s*\([^)]*\)", "", technology)

    technology_lower = technology.lower()

    aliases = {
        "react.js": "React",
        "reactjs": "React",
        "nodejs": "Node.js",
        "node.js": "Node.js",
        "html5": "HTML",
        "css3": "CSS"
    }

    if technology_lower in aliases:
        return aliases[technology_lower]

    known_technologies = {
        "python": "Python",
        "ast": "AST",
        "llms": "LLMs",
        "static code analysis": "Static Code Analysis",
        "react native": "React Native",
        "javascript": "JavaScript",
        "firebase auth": "Firebase Auth",
        "convex": "Convex",
        "mongodb": "MongoDB",
        "mysql": "MySQL",
        "sql": "SQL",
        "razorpay": "Razorpay",
        "twilio": "Twilio",
        "google maps api": "Google Maps API"
    }

    return known_technologies.get(
        technology_lower,
        technology
    )


for project in project_entries:

    normalized_technologies = []

    for technology in project["technologies"]:

        normalized = normalize_project_technology(technology)

        if normalized not in normalized_technologies:
            normalized_technologies.append(normalized)

    project["normalized_technologies"] = normalized_technologies


print("Normalized project technologies:\n")

for project in project_entries:

    print("=" * 60)
    print(project["name"])
    print("=" * 60)

    for technology in project["normalized_technologies"]:
        print("-", technology)
```

```python
# STEP 7.4 — Project ↔ JD Matching

jd_required_skills = jd_data["required_skills"]
jd_required_experience = jd_data["required_experience"]


def normalize_for_comparison(text):
    """
    Normalize text for safe exact comparison.
    """
    return text.strip().lower()


project_match_results = []


for project in project_entries:

    project_skills = project["normalized_technologies"]

    # ---------------------------------------------------------
    # 1. Match project technologies against JD skills
    # ---------------------------------------------------------

    matched_skills = []

    for project_skill in project_skills:

        project_skill_normalized = normalize_for_comparison(
            project_skill
        )

        for jd_skill in jd_required_skills:

            jd_skill_normalized = normalize_for_comparison(
                jd_skill
            )

            if project_skill_normalized == jd_skill_normalized:

                if jd_skill not in matched_skills:
                    matched_skills.append(jd_skill)


    # ---------------------------------------------------------
    # 2. Match project description against JD requirements
    # ---------------------------------------------------------

    description = project["description"].lower()

    matched_requirements = []


    # Full-stack development
    if (
        "full-stack" in description
        or "full stack" in description
    ):
        matched_requirements.append(
            "full stack development"
        )


    # Web application development
    if (
        "web platform" in description
        or "web application" in description
        or "web applications" in description
    ):
        matched_requirements.append(
            "web application development"
        )


    # Frontend development
    if (
        "frontend" in description
        or "front-end" in description
    ):
        matched_requirements.append(
            "frontend development"
        )


    # Backend development
    if (
        "backend" in description
        or "back-end" in description
    ):
        matched_requirements.append(
            "backend development"
        )


    # REST API development
    if (
        "rest api" in description
        or "rest apis" in description
    ):
        matched_requirements.append(
            "REST API development"
        )


    # Database development
    if (
        "database development" in description
        or "database" in description
    ):
        matched_requirements.append(
            "database development"
        )


    # Remove duplicates
    matched_requirements = list(
        dict.fromkeys(matched_requirements)
    )


    # ---------------------------------------------------------
    # 3. Calculate project scores
    # ---------------------------------------------------------

    skill_score = (
        len(matched_skills)
        / len(jd_required_skills)
        * 100
        if jd_required_skills
        else 0
    )


    requirement_score = (
        len(matched_requirements)
        / len(jd_required_experience)
        * 100
        if jd_required_experience
        else 0
    )


    # Project score combines both evidence sources
    project_score = (
        skill_score * 0.50
        + requirement_score * 0.50
    )


    # Project relevance classification
    if project_score >= 60:
        relevance = "HIGH"

    elif project_score >= 30:
        relevance = "MEDIUM"

    elif project_score > 0:
        relevance = "LOW"

    else:
        relevance = "NOT_RELEVANT"


    project_match_results.append({

        "name": project["name"],

        "matched_skills": matched_skills,

        "matched_requirements": matched_requirements,

        "skill_score": round(skill_score, 2),

        "requirement_score": round(
            requirement_score,
            2
        ),

        "project_score": round(
            project_score,
            2
        ),

        "relevance": relevance
    })


# ---------------------------------------------------------
# Display results
# ---------------------------------------------------------

print("Project ↔ JD Matching Results\n")


for result in project_match_results:

    print("=" * 60)
    print(result["name"])
    print("=" * 60)

    print("Matched JD skills:")

    if result["matched_skills"]:

        for skill in result["matched_skills"]:
            print("-", skill)

    else:
        print("- None")


    print("\nMatched JD requirements:")

    if result["matched_requirements"]:

        for requirement in result["matched_requirements"]:
            print("-", requirement)

    else:
        print("- None")


    print("\nSkill score       :", result["skill_score"], "%")
    print("Requirement score :", result["requirement_score"], "%")
    print("Project score     :", result["project_score"], "%")
    print("Relevance         :", result["relevance"])
```

```python
# STEP 10.1 — Final Candidate Evaluation

# ---------------------------------------------------------
# 1. Get component scores
# ---------------------------------------------------------

skill_score = skill_result["required_skill_score"]

experience_score = experience_match_result["experience_score"]

project_score = project_match_result["project_match_score"]


# ---------------------------------------------------------
# 2. Define weights
# ---------------------------------------------------------

weights = {
    "skills": 0.50,
    "experience": 0.30,
    "projects": 0.20
}


# ---------------------------------------------------------
# 3. Calculate weighted contributions
# ---------------------------------------------------------

skill_contribution = skill_score * weights["skills"]

experience_contribution = (
    experience_score * weights["experience"]
)

project_contribution = (
    project_score * weights["projects"]
)


# ---------------------------------------------------------
# 4. Calculate overall score
# ---------------------------------------------------------

overall_match_score = (
    skill_contribution
    + experience_contribution
    + project_contribution
)


# ---------------------------------------------------------
# 5. Determine overall status
# ---------------------------------------------------------

if overall_match_score >= 80:
    overall_status = "STRONG MATCH"

elif overall_match_score >= 60:
    overall_status = "GOOD MATCH"

elif overall_match_score >= 40:
    overall_status = "PARTIAL MATCH"

else:
    overall_status = "LOW MATCH"


# ---------------------------------------------------------
# 6. Build final structured result
# ---------------------------------------------------------

final_match_result = {

    "candidate_fit": {
        "overall_match_score": round(
            overall_match_score,
            2
        ),
        "overall_status": overall_status
    },

    "skills": {
        "matched_required_skills":
            skill_result["matched_required_skills"],

        "missing_required_skills":
            skill_result["missing_required_skills"],

        "matched_preferred_skills":
            skill_result["matched_preferred_skills"],

        "required_skill_score":
            skill_result["required_skill_score"]
    },

    "experience": {
        "required_months":
            experience_match_result["required_months"],

        "relevant_experience_months":
            experience_match_result[
                "relevant_experience_months"
            ],

        "experience_score":
            experience_match_result[
                "experience_score"
            ],

        "status":
            experience_match_result["status"],

        "technical_different_domain":
            experience_match_result[
                "technical_different_domain"
            ],

        "leadership_experience":
            experience_match_result[
                "leadership_experience"
            ]
    },

    "projects": {
        "total_projects":
            project_match_result["total_projects"],

        "relevant_projects":
            project_match_result["relevant_projects"],

        "high_relevance_projects":
            project_match_result["high_relevance_projects"],

        "medium_relevance_projects":
            project_match_result["medium_relevance_projects"],

        "low_relevance_projects":
            project_match_result["low_relevance_projects"],

        "not_relevant_projects":
            project_match_result["not_relevant_projects"],

        "project_match_score":
            project_match_result["project_match_score"]
    },

    "score_breakdown": {

        "skills": {
            "score": skill_score,
            "weight": weights["skills"],
            "contribution": round(
                skill_contribution,
                2
            )
        },

        "experience": {
            "score": experience_score,
            "weight": weights["experience"],
            "contribution": round(
                experience_contribution,
                2
            )
        },

        "projects": {
            "score": project_score,
            "weight": weights["projects"],
            "contribution": round(
                project_contribution,
                2
            )
        }
    }
}


# ---------------------------------------------------------
# 7. Display final result
# ---------------------------------------------------------

print("FINAL CANDIDATE EVALUATION")
print("=" * 60)

print(
    "Overall Match Score:",
    final_match_result["candidate_fit"][
        "overall_match_score"
    ],
    "%"
)

print(
    "Overall Status:",
    final_match_result["candidate_fit"][
        "overall_status"
    ]
)

print("\nScore Breakdown")
print("-" * 60)

print(
    "Skills:",
    skill_score,
    "% ×",
    weights["skills"],
    "=",
    skill_contribution
)

print(
    "Experience:",
    experience_score,
    "% ×",
    weights["experience"],
    "=",
    experience_contribution
)

print(
    "Projects:",
    project_score,
    "% ×",
    weights["projects"],
    "=",
    project_contribution
)

print("\nMatched Skills:")

for skill in skill_result["matched_required_skills"]:
    print("-", skill)

print("\nMissing Skills:")

for skill in skill_result["missing_required_skills"]:
    print("-", skill)
```

```python
# STEP 13.3 — Build Detailed Project Evidence
#
# This does NOT change the current project score.
# It creates a clearer evidence structure for the next stage.

project_evidence_for_questions = []

for project in project_entries:

    project_name = project["name"]

    # Normalize project technologies
    normalized_technologies = [
        normalize_project_technology(technology)
        for technology in project["technologies"]
    ]

    # Find exact JD skill matches
    matched_skills = sorted(
        set(normalized_technologies)
        & set(normalized_jd_skills)
    )

    # Find JD requirement evidence from project description
    description = project["description"].lower()

    matched_requirements = []

    for requirement in jd_data["required_experience"]:

        requirement_lower = requirement.lower()

        requirement_keywords = {
            "full stack development": [
                "full stack",
                "full-stack"
            ],
            "web application development": [
                "web platform",
                "web application",
                "web applications"
            ],
            "frontend development": [
                "frontend",
                "front-end"
            ],
            "backend development": [
                "backend",
                "back-end"
            ],
            "rest api development": [
                "rest api",
                "rest apis"
            ],
            "database development": [
                "database"
            ]
        }

        keywords = requirement_keywords.get(
            requirement_lower,
            [requirement_lower]
        )

        if any(
            keyword in description
            for keyword in keywords
        ):
            matched_requirements.append(requirement)

    # Determine evidence strength
    if matched_skills and matched_requirements:

        evidence_strength = "STRONG"

    elif matched_requirements:

        evidence_strength = "REQUIREMENT_EVIDENCE"

    elif matched_skills:

        evidence_strength = "TECHNOLOGY_EVIDENCE"

    else:

        evidence_strength = "NONE"

    # Determine project relevance
    if matched_skills and matched_requirements:

        relevance = "HIGH"

    elif matched_requirements or matched_skills:

        relevance = "LOW"

    else:

        relevance = "NOT_RELEVANT"

    project_evidence = {
        "project_name": project_name,
        "technologies": normalized_technologies,
        "matched_jd_skills": matched_skills,
        "matched_jd_requirements": matched_requirements,
        "missing_jd_skills": sorted(
            set(normalized_jd_skills)
            - set(matched_skills)
        ),
        "evidence_strength": evidence_strength,
        "relevance": relevance
    }

    project_evidence_for_questions.append(
        project_evidence
    )


# ---------------------------------------------------------
# Display the evidence
# ---------------------------------------------------------

print("DETAILED PROJECT EVIDENCE")
print("=" * 70)

for project in project_evidence_for_questions:

    print("\nProject:", project["project_name"])
    print("-" * 70)

    print("Technologies:")
    for technology in project["technologies"]:
        print("  -", technology)

    print("\nMatched JD skills:")
    if project["matched_jd_skills"]:
        for skill in project["matched_jd_skills"]:
            print("  -", skill)
    else:
        print("  None")

    print("\nMatched JD requirements:")
    if project["matched_jd_requirements"]:
        for requirement in project["matched_jd_requirements"]:
            print("  -", requirement)
    else:
        print("  None")

    print("\nMissing JD skills:")
    if project["missing_jd_skills"]:
        for skill in project["missing_jd_skills"]:
            print("  -", skill)
    else:
        print("  None")

    print(
        "\nEvidence strength:",
        project["evidence_strength"]
    )

    print(
        "Relevance:",
        project["relevance"]
    )
```

```python
# STEP 16G — Extract Text from Image-Based Resume Using OCR

import fitz
import pytesseract
from PIL import Image
import io

TEST_RESUME_PATH = DATA_DIR / "1785664313146_old_Resume_Bhavesh.pdf"

# Tell Python where Tesseract is installed
pytesseract.pytesseract.tesseract_cmd = (
    r"C:\Program Files\Tesseract-OCR\tesseract.exe"
)

# Open the PDF
doc = fitz.open(TEST_RESUME_PATH)

ocr_resume_text = ""

# Process every page
for page in doc:

    # Convert PDF page into an image
    pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))

    # Convert image bytes into a PIL image
    image = Image.open(io.BytesIO(pix.tobytes("png")))

    # Run OCR
    page_text = pytesseract.image_to_string(image)

    # Add the page text to the complete resume text
    ocr_resume_text += page_text + "\n"

doc.close()

print("OCR extraction completed!")
print("Characters:", len(ocr_resume_text))

print("\nExtracted Resume Text:\n")
print(ocr_resume_text[:3000])
```

```python
# ================================================================
# AI-POWERED RESUME–JD MATCHING AND CANDIDATE EVALUATION SYSTEM
# ================================================================
#
# VERSION:
# Corrected baseline
#
# FIXES INCLUDED:
#
# 1. PDF text extraction
# 2. OCR fallback
# 3. Safe OCR normalization
# 4. Resume section detection
# 5. Improved skill extraction
# 6. Parenthesis-aware skill parsing
# 7. Source-aware skill matching
# 8. Java != JavaScript
# 9. SQL != MySQL
# 10. React.js -> React
# 11. NodeJS -> Node.js
# 12. Improved experience evidence matching
# 13. Explicit experience-duration calculation
# 14. Correct project-title detection
# 15. Correct multi-line project grouping
# 16. Project relevance evaluation
# 17. Weighted candidate score
# 18. Structured JSON output
#
# IMPORTANT:
# This is currently a RULE-BASED / HEURISTIC baseline.
# It is NOT yet a trained ML model.
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

from pathlib import Path
import re
import json
import io
from datetime import date

import fitz
import pytesseract
from PIL import Image


# ================================================================
# 2. PATH CONFIGURATION
# ================================================================

BASE_DIR = Path("..")
DATA_DIR = BASE_DIR / "data"

# ------------------------------------------------
# CHANGE ONLY THIS FILE NAME WHEN TESTING ANOTHER
# RESUME.
# ------------------------------------------------

RESUME_FILENAME = "Sumit_JoshSoftware_resume.pdf"

RESUME_PATH = DATA_DIR / RESUME_FILENAME


print("=" * 80)
print("RESUME–JD MATCHING SYSTEM")
print("=" * 80)

print("\nResume path:")
print(RESUME_PATH)

print("\nFile exists:", RESUME_PATH.exists())

if not RESUME_PATH.exists():

    raise FileNotFoundError(
        f"\nResume not found:\n{RESUME_PATH}\n"
        f"\nPlease check the filename in the data folder."
    )


# ================================================================
# 3. TESSERACT CONFIGURATION
# ================================================================

TESSERACT_PATH = (
    r"C:\Program Files\Tesseract-OCR\tesseract.exe"
)

pytesseract.pytesseract.tesseract_cmd = (
    TESSERACT_PATH
)

try:

    print(
        "\nTesseract version:",
        pytesseract.get_tesseract_version()
    )

except Exception:

    print(
        "\nWarning: Tesseract could not be verified."
    )


# ================================================================
# 4. EXTRACT SELECTABLE PDF TEXT
# ================================================================

doc = fitz.open(
    RESUME_PATH
)

selectable_text = ""

for page in doc:

    selectable_text += page.get_text()

page_count = len(doc)

doc.close()


print(
    "\nSelectable PDF text characters:",
    len(selectable_text)
)

print(
    "PDF pages:",
    page_count
)


# ================================================================
# 5. OCR FALLBACK
# ================================================================

OCR_MIN_CHARACTERS = 100


if len(selectable_text.strip()) >= OCR_MIN_CHARACTERS:

    resume_text = selectable_text

    print(
        "\nSelectable text detected."
    )

    print(
        "OCR not required."
    )

else:

    print(
        "\nSelectable text insufficient."
    )

    print(
        "Running OCR..."
    )

    doc = fitz.open(
        RESUME_PATH
    )

    ocr_pages = []

    for page_number, page in enumerate(
        doc,
        start=1
    ):

        pix = page.get_pixmap(
            matrix=fitz.Matrix(2, 2),
            alpha=False
        )

        image = Image.open(
            io.BytesIO(
                pix.tobytes("png")
            )
        )

        page_text = pytesseract.image_to_string(
            image
        )

        ocr_pages.append(
            page_text
        )

        print(
            f"OCR completed: "
            f"{page_number}/{len(doc)}"
        )

    doc.close()

    resume_text = "\n".join(
        ocr_pages
    )

    print(
        "\nOCR extraction completed."
    )


# ================================================================
# 6. GENERAL TEXT CLEANUP
# ================================================================

def clean_resume_text(text):

    # Normalize Windows line endings.
    text = text.replace(
        "\r\n",
        "\n"
    )

    text = text.replace(
        "\r",
        "\n"
    )

    # ------------------------------------------------------------
    # Fix words split across PDF lines:
    #
    # de-
    # ployed
    #
    # becomes:
    #
    # deployed
    # ------------------------------------------------------------

    text = re.sub(
        r"(\w)-\s*\n\s*(\w)",
        r"\1\2",
        text
    )

    # Normalize spaces but preserve newlines.
    text = re.sub(
        r"[ \t]+",
        " ",
        text
    )

    # Remove excessive blank lines.
    text = re.sub(
        r"\n{3,}",
        "\n\n",
        text
    )

    return text.strip()


resume_text = clean_resume_text(
    resume_text
)


# ================================================================
# 7. CONTROLLED OCR NORMALIZATION
# ================================================================

def controlled_ocr_normalization(text):

    replacements = {

        "APls": "APIs",
        "APl": "API",

        "Al-driven": "AI-driven",
        "Al-powered": "AI-powered",

        "Gemini Al": "Gemini AI",
    }

    for old, new in replacements.items():

        text = text.replace(
            old,
            new
        )

    return text


resume_text = controlled_ocr_normalization(
    resume_text
)


print(
    "\nFinal extracted characters:",
    len(resume_text)
)


# ================================================================
# 8. SECTION DEFINITIONS
# ================================================================

SECTION_ALIASES = {

    "Professional Summary": [

        "Professional Summary",
        "Summary",
        "Profile",
        "About Me",
        "Objective"
    ],

    "Skills": [

        "Technical Skills",
        "Technical Skill",
        "Skills",
        "Core Skills",
        "Key Skills",
        "Technologies",
        "Technical Expertise"
    ],

    "Internship & Experience": [

        "Professional Experience",
        "Experience",
        "Work Experience",
        "Internship",
        "Internships",
        "Internship & Experience",
        "Employment History",
        "Career Experience"
    ],

    "Projects": [

        "Projects",
        "Selected Projects",
        "Academic Projects",
        "Personal Projects",
        "Project Experience",
        "Projects & Research"
    ],

    "Education": [

        "Education",
        "Academic Background",
        "Educational Qualification",
        "Educational Qualifications"
    ],

    "Certifications": [

        "Certifications",
        "Certificates",
        "Professional Certifications"
    ],

    "Achievements": [

        "Achievements",
        "Awards",
        "Honors"
    ],

    "Leadership & Volunteering": [

        "Leadership & Volunteering",
        "Leadership",
        "Volunteering",
        "Positions of Responsibility"
    ]
}


# ================================================================
# 9. HEADING NORMALIZATION
# ================================================================

def normalize_heading(line):

    line = line.strip()

    line = re.sub(
        r"^[•●▪◦*\-]+\s*",
        "",
        line
    )

    line = re.sub(
        r"\s+",
        " ",
        line
    )

    return line.strip().lower()


HEADING_LOOKUP = {}

for section_name, aliases in SECTION_ALIASES.items():

    for alias in aliases:

        HEADING_LOOKUP[
            normalize_heading(alias)
        ] = section_name


# ================================================================
# 10. SECTION DETECTION
# ================================================================

def detect_sections(text):

    lines = text.splitlines()

    sections = {}

    current_section = None

    for raw_line in lines:

        line = raw_line.strip()

        if not line:

            continue

        normalized = normalize_heading(
            line
        )

        if normalized in HEADING_LOOKUP:

            current_section = HEADING_LOOKUP[
                normalized
            ]

            if current_section not in sections:

                sections[current_section] = []

            continue

        if current_section is not None:

            sections[current_section].append(
                line
            )

    return sections


sections = detect_sections(
    resume_text
)


# ================================================================
# 11. DISPLAY DETECTED SECTIONS
# ================================================================

print("\n" + "=" * 80)
print("DETECTED RESUME SECTIONS")
print("=" * 80)

for section_name, lines in sections.items():

    print(
        f"\n{section_name}: "
        f"{len(lines)} lines"
    )


# ================================================================
# 12. NAME EXTRACTION
# ================================================================

def extract_name(text):

    lines = text.splitlines()

    for line in lines:

        line = line.strip()

        if not line:
            continue

        normalized = normalize_heading(
            line
        )

        if normalized in HEADING_LOOKUP:
            continue

        if "@" in line:
            continue

        if re.search(
            r"\d{7,}",
            line
        ):
            continue

        if len(line.split()) > 5:
            continue

        if re.fullmatch(
            r"[A-Za-z][A-Za-z .'-]{1,60}",
            line
        ):

            return line

    return "Unknown"


candidate_name = extract_name(
    resume_text
)


# ================================================================
# 13. PHONE EXTRACTION
# ================================================================

phone_match = re.search(
    r"(?<!\d)(?:\+?\d[\d\s().-]{8,}\d)(?!\d)",
    resume_text
)

candidate_phone = (

    phone_match.group(0).strip()

    if phone_match

    else "Unknown"
)


# ================================================================
# 14. EMAIL EXTRACTION
# ================================================================

email_match = re.search(
    r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
    resume_text
)

candidate_email = (

    email_match.group(0)

    if email_match

    else "Unknown"
)


# ================================================================
# 15. JOB DESCRIPTION
# ================================================================

jd_data = {

    "job_title":
        "Junior Full Stack Developer",

    "experience": {

        "min_years": 1,

        "max_years": None
    },

    "required_experience": [

        "full stack development",
        "web application development",
        "frontend development",
        "backend development",
        "REST API development",
        "database development"
    ],

    "required_skills": [

        "Java",
        "Spring Boot",
        "MySQL",
        "HTML",
        "CSS",
        "JavaScript",
        "React",
        "Node.js"
    ],

    "preferred_skills": []
}


# ================================================================
# 16. SKILL NORMALIZATION
# ================================================================

def normalize_skill(skill):

    skill = skill.strip()

    skill = re.sub(
        r"^[•●▪◦*\-]+\s*",
        "",
        skill
    )

    skill = re.sub(
        r"\s+",
        " ",
        skill
    )

    skill = skill.strip(
        " ,;:"
    )

    normalized = skill.lower()

    aliases = {

        "react.js": "React",
        "reactjs": "React",

        "nodejs": "Node.js",
        "node.js": "Node.js",

        "html5": "HTML",
        "html": "HTML",

        "css3": "CSS",
        "css": "CSS",

        "springboot": "Spring Boot",
        "spring boot": "Spring Boot",

        "mysql": "MySQL",

        "mongodb": "MongoDB",

        "sql": "SQL",

        "express": "Express.js",
        "expressjs": "Express.js",
        "express.js": "Express.js",

        "c++": "C++",

        "github": "GitHub",
        "gitlab": "GitLab"
    }

    return aliases.get(
        normalized,
        skill
    )


# ================================================================
# 17. TOP-LEVEL COMMA SPLITTER
# ================================================================

def split_top_level_commas(text):

    parts = []

    current = []

    depth = 0

    for char in text:

        if char == "(":

            depth += 1

        elif char == ")":

            depth = max(
                0,
                depth - 1
            )

        if char == "," and depth == 0:

            part = "".join(
                current
            ).strip()

            if part:

                parts.append(
                    part
                )

            current = []

        else:

            current.append(
                char
            )

    final_part = "".join(
        current
    ).strip()

    if final_part:

        parts.append(
            final_part
        )

    return parts


# ================================================================
# 18. CANDIDATE SKILL EXTRACTION
# ================================================================

def extract_candidate_skills(
    skills_lines
):

    extracted = []

    for line in skills_lines:

        line = line.strip()

        if not line:
            continue

        line = re.sub(
            r"^[•●▪◦*\-]+\s*",
            "",
            line
        )

        # Remove category name.
        if ":" in line:

            _, values = line.split(
                ":",
                1
            )

            values = values.strip()

        else:

            values = line

        items = split_top_level_commas(
            values
        )

        for item in items:

            item = item.strip()

            if not item:
                continue

            # ----------------------------------------------------
            # Parenthetical group:
            #
            # AWS (Lambda, Amplify, S3, Bedrock, SES, EC2)
            # ----------------------------------------------------

            parenthetical = re.fullmatch(
                r"(.+?)\s*\(([^()]*)\)",
                item
            )

            if parenthetical:

                main_skill = parenthetical.group(
                    1
                ).strip()

                sub_skills = parenthetical.group(
                    2
                ).strip()

                normalized_main = normalize_skill(
                    main_skill
                )

                if normalized_main:

                    extracted.append(
                        normalized_main
                    )

                for sub_skill in split_top_level_commas(
                    sub_skills
                ):

                    normalized_sub = normalize_skill(
                        sub_skill
                    )

                    if normalized_sub:

                        extracted.append(
                            normalized_sub
                        )

            else:

                normalized = normalize_skill(
                    item
                )

                if normalized:

                    extracted.append(
                        normalized
                    )

    # ------------------------------------------------------------
    # Remove duplicates.
    # ------------------------------------------------------------

    unique = []

    seen = set()

    for skill in extracted:

        key = skill.lower()

        if key not in seen:

            seen.add(
                key
            )

            unique.append(
                skill
            )

    return unique


candidate_skills = extract_candidate_skills(
    sections.get(
        "Skills",
        []
    )
)


# ================================================================
# 19. NORMALIZE JD SKILLS
# ================================================================

normalized_jd_skills = [

    normalize_skill(skill)

    for skill in jd_data[
        "required_skills"
    ]
]


# ================================================================
# 20. REQUIRED SKILL REGEX PATTERNS
# ================================================================
#
# IMPORTANT:
#
# We intentionally DO NOT use:
#
# "js" -> JavaScript
#
# because "Node.js" contains "js".
#
# This was the source of the false JavaScript evidence.
# ================================================================

SKILL_PATTERNS = {

    "Java": [

        r"(?<!\w)java(?!script)(?!\w)"
    ],

    "Spring Boot": [

        r"(?<!\w)spring\s*boot(?!\w)",

        r"(?<!\w)springboot(?!\w)"
    ],

    "MySQL": [

        r"(?<!\w)mysql(?!\w)"
    ],

    "HTML": [

        r"(?<!\w)html(?:5)?(?!\w)"
    ],

    "CSS": [

        r"(?<!\w)css(?:3)?(?!\w)"
    ],

    "JavaScript": [

        r"(?<!\w)javascript(?!\w)"
    ],

    "React": [

        r"(?<!\w)react(?:\.js|js)?(?!\w)"
    ],

    "Node.js": [

        r"(?<!\w)node(?:\.js|js)(?!\w)"
    ]
}


# ================================================================
# 21. SKILL EVIDENCE FINDER
# ================================================================

def find_skill_evidence(
    skill,
    text
):

    patterns = SKILL_PATTERNS.get(
        skill,
        [
            r"(?<!\w)"
            + re.escape(skill)
            + r"(?!\w)"
        ]
    )

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            re.IGNORECASE
        )

        if match:

            start = max(
                0,
                match.start() - 50
            )

            end = min(
                len(text),
                match.end() + 50
            )

            context = text[
                start:end
            ].strip()

            return {

                "matched": True,

                "matched_text":
                    match.group(0),

                "context":
                    context
            }

    return {

        "matched": False,

        "matched_text": None,

        "context": None
    }


# ================================================================
# 22. SOURCE-AWARE REQUIRED SKILL MATCHING
# ================================================================

SOURCE_PRIORITY = [

    "Skills",
    "Internship & Experience",
    "Projects",
    "Professional Summary"
]


def match_required_skills_with_sources(
    jd_skills,
    sections
):

    matched = []

    missing = []

    evidence = {}

    for required_skill in jd_skills:

        skill_sources = []

        for source in SOURCE_PRIORITY:

            source_lines = sections.get(
                source,
                []
            )

            source_text = "\n".join(
                source_lines
            )

            result = find_skill_evidence(
                required_skill,
                source_text
            )

            if result["matched"]:

                skill_sources.append({

                    "source": source,

                    "matched_text":
                        result[
                            "matched_text"
                        ],

                    "context":
                        result[
                            "context"
                        ]
                })

        if skill_sources:

            matched.append(
                required_skill
            )

            evidence[
                required_skill
            ] = skill_sources

        else:

            missing.append(
                required_skill
            )

            evidence[
                required_skill
            ] = []

    score = (

        len(matched)
        /
        len(jd_skills)
        * 100

        if jd_skills

        else 0
    )

    return (

        matched,

        missing,

        evidence,

        round(
            score,
            2
        )
    )


(
    matched_required_skills,
    missing_required_skills,
    skill_evidence,
    skill_score
) = match_required_skills_with_sources(
    normalized_jd_skills,
    sections
)


# ================================================================
# 23. EXPERIENCE EVIDENCE RULES
# ================================================================

experience_evidence_rules = {

    "full stack development": [

        r"\bfull[- ]stack\b",

        r"\bend[- ]to[- ]end\b",

        r"\bmern\b",

        r"\bmean\b",

        r"frontend\s+and\s+backend"
    ],

    "web application development": [

        r"\bweb[- ]based\s+applications?\b",

        r"\bweb\s+applications?\b",

        r"\bweb\s+platform\b",

        r"\bdynamic\s+web\b",

        r"\bcontent\s+management\s+panel\b"
    ],

    "frontend development": [

        r"\bfrontend\b",

        r"\bfront[- ]end\b",

        r"\buser\s+interface\b",

        r"\bui\b"
    ],

    "backend development": [

        r"\bbackend\b",

        r"\bback[- ]end\b",

        r"\bserver[- ]side\b",

        r"\bserver\s+side\b"
    ],

    "REST API development": [

        r"\bRESTful?\s+API(?:s)?\b",

        r"\bREST\s+API(?:s)?\b",

        r"\bAPI\s+development\b",

        r"\bcustom\s+.*?\bAPI(?:s)?\b",

        r"\bAPI\s+System\b",

        r"\bAPI(?:s)?\b"
    ],

    "database development": [

        r"\bdatabase\b",

        r"\bdatabases\b",

        r"\bdatabase\s+queries\b",

        r"\brelational\s+database\b",

        r"\bMongoDB\b",

        r"\bMySQL\b",

        r"\bSQL\b",

        r"\bFirebase\b"
    ]
}


# ================================================================
# 24. STRONG EXPERIENCE EVIDENCE
# ================================================================

experience_strong_patterns = {

    "full stack development": [

        r"\bfull[- ]stack\b",

        r"\bend[- ]to[- ]end\b",

        r"\bmern\b",

        r"\bmean\b"
    ],

    "web application development": [

        r"\bweb[- ]based\s+applications?\b",

        r"\bweb\s+applications?\b",

        r"\bweb\s+platform\b",

        r"\bcontent\s+management\s+panel\b"
    ],

    "frontend development": [

        r"\bfrontend\b",

        r"\bfront[- ]end\b",

        r"\buser\s+interface\b",

        r"\bui\b"
    ],

    "backend development": [

        r"\bbackend\b",

        r"\bback[- ]end\b",

        r"\bserver[- ]side\b",

        r"\bserver\s+side\b"
    ],

    "REST API development": [

        r"\bRESTful?\s+API(?:s)?\b",

        r"\bREST\s+API(?:s)?\b",

        r"\bAPI\s+development\b",

        r"\bcustom\s+.*?\bAPI(?:s)?\b"
    ],

    "database development": [

        r"\bdatabase\b",

        r"\bdatabases\b",

        r"\bdatabase\s+queries\b",

        r"\brelational\s+database\b"
    ]
}


# ================================================================
# 25. EXPERIENCE EVALUATION
# ================================================================

def evaluate_experience_evidence(
    experience_text
):

    results = {}

    total_score = 0

    for requirement, patterns in (
        experience_evidence_rules.items()
    ):

        found = []

        # --------------------------------------------------------
        # Search every evidence pattern.
        # --------------------------------------------------------

        for pattern in patterns:

            matches = re.finditer(
                pattern,
                experience_text,
                re.IGNORECASE
            )

            for match in matches:

                matched_text = match.group(
                    0
                ).strip()

                if matched_text not in found:

                    found.append(
                        matched_text
                    )

        # --------------------------------------------------------
        # Determine whether evidence is strong.
        # --------------------------------------------------------

        strong_patterns = experience_strong_patterns.get(
            requirement,
            []
        )

        strong_found = []

        for pattern in strong_patterns:

            if re.search(
                pattern,
                experience_text,
                re.IGNORECASE
            ):

                match = re.search(
                    pattern,
                    experience_text,
                    re.IGNORECASE
                )

                if match:

                    strong_found.append(
                        match.group(0)
                    )

        if strong_found:

            strength = "STRONG"

            numeric_score = 1.0

        elif found:

            strength = "PARTIAL"

            numeric_score = 0.5

        else:

            strength = "NONE"

            numeric_score = 0.0

        total_score += numeric_score

        results[
            requirement
        ] = {

            "evidence":
                found,

            "strength":
                strength,

            "numeric_score":
                numeric_score
        }

    overall_score = (

        total_score
        /
        len(
            experience_evidence_rules
        )
        * 100

        if experience_evidence_rules

        else 0
    )

    return (

        results,

        round(
            overall_score,
            2
        )
    )


experience_text = "\n".join(
    sections.get(
        "Internship & Experience",
        []
    )
)


experience_results, experience_score = (
    evaluate_experience_evidence(
        experience_text
    )
)


# ================================================================
# 26. MONTH/YEAR PARSING
# ================================================================

MONTH_MAP = {

    "jan": 1,
    "january": 1,

    "feb": 2,
    "february": 2,

    "mar": 3,
    "march": 3,

    "apr": 4,
    "april": 4,

    "may": 5,

    "jun": 6,
    "june": 6,

    "jul": 7,
    "july": 7,

    "aug": 8,
    "august": 8,

    "sep": 9,
    "sept": 9,
    "september": 9,

    "oct": 10,
    "october": 10,

    "nov": 11,
    "november": 11,

    "dec": 12,
    "december": 12
}


# ================================================================
# 27. PARSE MONTH/YEAR
# ================================================================

def parse_month_year(text):

    text = text.strip().lower()

    match = re.search(
        r"\b("
        r"jan(?:uary)?|"
        r"feb(?:ruary)?|"
        r"mar(?:ch)?|"
        r"apr(?:il)?|"
        r"may|"
        r"jun(?:e)?|"
        r"jul(?:y)?|"
        r"aug(?:ust)?|"
        r"sep(?:t(?:ember)?)?|"
        r"oct(?:ober)?|"
        r"nov(?:ember)?|"
        r"dec(?:ember)?"
        r")"
        r"\s+"
        r"(\d{4})",
        text,
        re.IGNORECASE
    )

    if not match:

        return None

    month_text = match.group(
        1
    ).lower()

    year = int(
        match.group(2)
    )

    month = MONTH_MAP.get(
        month_text
    )

    if not month:

        return None

    return date(
        year,
        month,
        1
    )


# ================================================================
# 28. EXPERIENCE DATE RANGE EXTRACTION
# ================================================================

def extract_experience_ranges(
    experience_text
):

    pattern = re.compile(

        r"("
        r"(?:Jan(?:uary)?|"
        r"Feb(?:ruary)?|"
        r"Mar(?:ch)?|"
        r"Apr(?:il)?|"
        r"May|"
        r"Jun(?:e)?|"
        r"Jul(?:y)?|"
        r"Aug(?:ust)?|"
        r"Sep(?:t(?:ember)?)?|"
        r"Oct(?:ober)?|"
        r"Nov(?:ember)?|"
        r"Dec(?:ember)?)"
        r"\s+\d{4}"
        r")"
        r"\s*[–—-]\s*"
        r"("
        r"(?:Jan(?:uary)?|"
        r"Feb(?:ruary)?|"
        r"Mar(?:ch)?|"
        r"Apr(?:il)?|"
        r"May|"
        r"Jun(?:e)?|"
        r"Jul(?:y)?|"
        r"Aug(?:ust)?|"
        r"Sep(?:t(?:ember)?)?|"
        r"Oct(?:ober)?|"
        r"Nov(?:ember)?|"
        r"Dec(?:ember)?)"
        r"\s+\d{4}"
        r")",

        re.IGNORECASE
    )

    ranges = []

    for match in pattern.finditer(
        experience_text
    ):

        start_date = parse_month_year(
            match.group(1)
        )

        end_date = parse_month_year(
            match.group(2)
        )

        if start_date and end_date:

            ranges.append({

                "start":
                    match.group(1),

                "end":
                    match.group(2),

                "start_date":
                    start_date,

                "end_date":
                    end_date
            })

    return ranges


# ================================================================
# 29. INCLUSIVE EXPERIENCE MONTH CALCULATION
# ================================================================
#
# Example:
#
# Sep 2025 -> Jun 2026
#
# September, October, November, December,
# January, February, March, April, May, June
#
# = 10 months
#
# Therefore we use:
#
# (end.year-start.year)*12
# + (end.month-start.month)
# + 1
#
# ================================================================

def calculate_inclusive_months(
    start_date,
    end_date
):

    if end_date < start_date:

        return 0

    return (

        (end_date.year - start_date.year)
        * 12

        +

        (end_date.month - start_date.month)

        + 1
    )


# ================================================================
# 30. EXPERIENCE DURATION
# ================================================================

def extract_experience_duration(
    experience_text
):

    ranges = extract_experience_ranges(
        experience_text
    )

    if not ranges:

        return {

            "status":
                "UNKNOWN",

            "total_months":
                None,

            "total_years":
                None,

            "ranges":
                []
        }

    total_months = 0

    output_ranges = []

    for item in ranges:

        months = calculate_inclusive_months(
            item["start_date"],
            item["end_date"]
        )

        total_months += months

        output_ranges.append({

            "start":
                item["start"],

            "end":
                item["end"],

            "months":
                months
        })

    return {

        "status":
            "CALCULATED_FROM_EXPLICIT_RANGES",

        "total_months":
            total_months,

        "total_years":
            round(
                total_months / 12,
                2
            ),

        "ranges":
            output_ranges
    }


experience_duration = extract_experience_duration(
    experience_text
)


# ================================================================
# 31. EXPERIENCE ELIGIBILITY
# ================================================================

required_experience_years = jd_data[
    "experience"
]["min_years"]


if experience_duration[
    "total_years"
] is None:

    experience_eligibility = (
        "UNKNOWN"
    )

elif experience_duration[
    "total_years"
] >= required_experience_years:

    experience_eligibility = (
        "MEETS_REQUIREMENT"
    )

else:

    experience_eligibility = (
        "BELOW_REQUIREMENT"
    )


# ================================================================
# 32. PROJECT DETECTION HELPERS
# ================================================================

# ================================================================
# 32. PROJECT DETECTION HELPERS
# ================================================================

def is_bullet_line(line):
    """
    Check whether a resume line starts with a bullet.
    """
    return bool(
        re.match(
            r"^\s*[•●▪◦*]\s*",
            line
        )
    )


def clean_project_line(line):
    """
    Remove bullet symbols from the beginning of a line.
    """
    line = line.strip()

    line = re.sub(
        r"^\s*[•●▪◦*]\s*",
        "",
        line
    )

    return line.strip()


# ================================================================
# 33. PROJECT TITLE DETECTION
# ================================================================

def split_project_title_and_description(line):
    """
    Detect project entries written like:

    • AICoRex – AI-Driven Continuous Codebase Evaluation System
    • DietGenius – AI Diet Planner Android App
    • VIKRSHI – Farm Tool Renting System

    Returns:
        (title, description)

    If the line is not a project title, returns:
        (None, None)
    """

    cleaned = clean_project_line(line)

    if not cleaned:
        return None, None

    # ------------------------------------------------------------
    # Project titles commonly use an em dash or en dash.
    # ------------------------------------------------------------

    dash_match = re.split(
        r"\s+[–—]\s+|\s+[–—]\s*",
        cleaned,
        maxsplit=1
    )

    if len(dash_match) == 2:

        title = dash_match[0].strip()
        description = dash_match[1].strip()

        if title and len(title) <= 100:

            return title, description

    # ------------------------------------------------------------
    # Also support:
    #
    # Project Name - Description
    # ------------------------------------------------------------

    hyphen_match = re.split(
        r"\s+-\s+",
        cleaned,
        maxsplit=1
    )

    if len(hyphen_match) == 2:

        title = hyphen_match[0].strip()
        description = hyphen_match[1].strip()

        if title and len(title) <= 100:

            return title, description

    return None, None


def looks_like_standalone_project_title(line):
    """
    Handles project titles that appear on their own line.
    """

    cleaned = clean_project_line(line)

    if not cleaned:
        return False

    lower_line = cleaned.lower()

    # Never treat obvious descriptions as titles.
    description_starts = (
        "built ",
        "developed ",
        "designed ",
        "created ",
        "implemented ",
        "deployed ",
        "integrated ",
        "used ",
        "leveraged ",
        "improved ",
        "worked ",
        "participated "
    )

    if lower_line.startswith(description_starts):
        return False

    # Ignore very long sentences.
    if len(cleaned) > 100:
        return False

    # Project-like keywords.
    project_keywords = (
        "system",
        "platform",
        "application",
        "app",
        "engine",
        "portal",
        "website",
        "management"
    )

    if any(
        keyword in lower_line
        for keyword in project_keywords
    ):
        return True

    # Title-case heuristic.
    words = re.findall(
        r"[A-Za-z][A-Za-z'-]*",
        cleaned
    )

    if not words:
        return False

    capitalized = sum(
        1
        for word in words
        if word[0].isupper()
    )

    ratio = capitalized / len(words)

    return ratio >= 0.60


# ================================================================
# 34. PROJECT EXTRACTION
# ================================================================

def extract_projects(project_lines):

    projects = []

    current_project = None

    for raw_line in project_lines:

        line = raw_line.strip()

        if not line:
            continue

        # --------------------------------------------------------
        # FIRST:
        # Check whether this bullet itself contains:
        #
        # Project Title – Description
        # --------------------------------------------------------

        title, description = (
            split_project_title_and_description(line)
        )

        if title:

            # Save previous project.
            if current_project is not None:

                current_project["description"] = (
                    " ".join(
                        current_project[
                            "description_lines"
                        ]
                    ).strip()
                )

                projects.append(
                    current_project
                )

            # Start new project.
            current_project = {

                "title": title,

                "description_lines": []

            }

            # Add same-line description.
            if description:

                current_project[
                    "description_lines"
                ].append(
                    description
                )

            continue

        # --------------------------------------------------------
        # SECOND:
        # Standalone project title.
        # --------------------------------------------------------

        if (
            not is_bullet_line(line)
            and
            looks_like_standalone_project_title(line)
        ):

            # Save previous project.
            if current_project is not None:

                current_project["description"] = (
                    " ".join(
                        current_project[
                            "description_lines"
                        ]
                    ).strip()
                )

                projects.append(
                    current_project
                )

            # Start new project.
            current_project = {

                "title":
                    clean_project_line(line),

                "description_lines":
                    []

            }

            continue

        # --------------------------------------------------------
        # THIRD:
        # Any remaining line belongs to current project.
        # --------------------------------------------------------

        if current_project is not None:

            cleaned_line = clean_project_line(
                line
            )

            if cleaned_line:

                current_project[
                    "description_lines"
                ].append(
                    cleaned_line
                )

    # ------------------------------------------------------------
    # Save final project.
    # ------------------------------------------------------------

    if current_project is not None:

        current_project["description"] = (
            " ".join(
                current_project[
                    "description_lines"
                ]
            ).strip()
        )

        projects.append(
            current_project
        )

    return projects


# ================================================================
# RUN PROJECT EXTRACTION
# ================================================================

projects = extract_projects(
    sections.get(
        "Projects",
        []
    )
)

# ================================================================
# 35. PROJECT EVIDENCE RULES
# ================================================================

project_evidence_rules = {

    "full stack development": [

        r"\bfull[- ]stack\b",

        r"\bmern\b",

        r"\bmean\b",

        r"\bend[- ]to[- ]end\b",

        r"\bfrontend\s+and\s+backend\b"

    ],

    "web application development": [

        r"\bweb\s+application\b",

        r"\bweb\s+applications\b",

        r"\bweb[- ]based\s+application\b",

        r"\bweb[- ]based\s+applications\b",

        r"\bweb\s+platform\b",

        r"\bweb[- ]platform\b",

        r"\bweb\s+portal\b",

        r"\bwebsite\b",

        r"\bdynamic\s+web\b"

    ],

    "frontend development": [

        r"\breact(?:\.js)?\b",

        r"\breact[- ]native\b",

        r"\breact-based\b",

        r"\bfrontend\b",

        r"\bfront[- ]end\b",

        r"\buser\s+interface\b",

        r"\bUI\b"

    ],

    "backend development": [

        r"\bbackend\b",

        r"\bback[- ]end\b",

        r"\bserver[- ]side\b",

        r"\bserver\s+side\b",

        r"\bnode(?:\.js|js)\b",

        r"\bPHP\b",

        r"\bpython\b"

    ],

    "REST API development": [

        r"\bRESTful?\s+API(?:s)?\b",

        r"\bREST\s+API(?:s)?\b",

        r"\bcustom\s+.*?\bAPI(?:s)?\b",

        r"\bAPI\s+development\b",

        r"\bAPI\s+integration\b",

        r"\bAPI\s+integrations\b",

        r"\bAPI\b"

    ],

    "database development": [

        r"\bdatabase\b",

        r"\bdatabases\b",

        r"\bMongoDB\b",

        r"\bMySQL\b",

        r"\bSQL\b",

        r"\bFirebase\b",

        r"\bFirestore\b",

        r"\bPostgreSQL\b",

        r"\bConvex\b"
    ]
}

# ================================================================
# 36. PROJECT EVALUATION
# ================================================================

def evaluate_project(
    project
):

    title = project[
        "title"
    ]

    description = project[
        "description"
    ]

    searchable_text = (
        title
        + " "
        + description
    )

    evidence_result = {}

    total_score = 0

    for requirement, patterns in (
        project_evidence_rules.items()
    ):

        found = []

        for pattern in patterns:

            matches = re.finditer(
                pattern,
                searchable_text,
                re.IGNORECASE
            )

            for match in matches:

                matched = match.group(
                    0
                ).strip()

                if matched not in found:

                    found.append(
                        matched
                    )

        # --------------------------------------------------------
        # Full-stack requires explicit evidence.
        # --------------------------------------------------------

        if requirement == (
            "full stack development"
        ):

            strong_patterns = [

                r"\bfull[- ]stack\b",
                r"\bmern\b",
                r"\bmean\b",
                r"\bend[- ]to[- ]end\b"
            ]

            strong = any(
                re.search(
                    pattern,
                    searchable_text,
                    re.IGNORECASE
                )
                for pattern
                in strong_patterns
            )

            if strong:

                strength = "STRONG"
                numeric_score = 1.0

            else:

                found = []
                strength = "NONE"
                numeric_score = 0.0

        # --------------------------------------------------------
        # REST API:
        # RESTful API = strong
        # Generic API = partial
        # --------------------------------------------------------

        elif requirement == (
            "REST API development"
        ):

            strong_patterns = [

                r"\bRESTful?\s+API(?:s)?\b",

                r"\bREST\s+API(?:s)?\b",

                r"\bcustom\s+.*?\bAPI(?:s)?\b"
            ]

            strong = any(
                re.search(
                    pattern,
                    searchable_text,
                    re.IGNORECASE
                )
                for pattern
                in strong_patterns
            )

            if strong:

                strength = "STRONG"
                numeric_score = 1.0

            elif found:

                strength = "PARTIAL"
                numeric_score = 0.5

            else:

                strength = "NONE"
                numeric_score = 0.0

        else:

            if found:

                strength = "STRONG"
                numeric_score = 1.0

            else:

                strength = "NONE"
                numeric_score = 0.0

        total_score += numeric_score

        evidence_result[
            requirement
        ] = {

            "evidence":
                found,

            "strength":
                strength,

            "numeric_score":
                numeric_score
        }

    project_score = (

        total_score
        /
        len(project_evidence_rules)
        * 100

        if project_evidence_rules

        else 0
    )

    return {

        "title":
            title,

        "description":
            description,

        "evidence":
            evidence_result,

        "project_match_score":
            round(
                project_score,
                2
            )
    }


# ================================================================
# 37. EVALUATE ALL PROJECTS
# ================================================================

project_results = []

for project in projects:

    result = evaluate_project(
        project
    )

    project_results.append(
        result
    )


# ================================================================
# 38. PROJECT AGGREGATE SCORE
# ================================================================

if project_results:

    project_score = round(

        sum(
            project[
                "project_match_score"
            ]

            for project
            in project_results
        )

        /

        len(project_results),

        2
    )

else:

    project_score = 0.0


# ================================================================
# 39. OVERALL SCORE
# ================================================================

SKILL_WEIGHT = 0.50

EXPERIENCE_WEIGHT = 0.30

PROJECT_WEIGHT = 0.20


overall_score = (

    skill_score
    * SKILL_WEIGHT

    +

    experience_score
    * EXPERIENCE_WEIGHT

    +

    project_score
    * PROJECT_WEIGHT
)


overall_score = round(
    overall_score,
    2
)


# ================================================================
# 40. STATUS
# ================================================================

def score_status(
    score
):

    if score >= 70:

        return "HIGH MATCH"

    elif score >= 40:

        return "MEDIUM MATCH"

    else:

        return "LOW MATCH"


overall_status = score_status(
    overall_score
)


# ================================================================
# 41. STRUCTURED JSON
# ================================================================

evaluation_json = {

    "candidate": {

        "name":
            candidate_name,

        "phone":
            candidate_phone,

        "email":
            candidate_email
    },

    "job": {

        "job_title":
            jd_data[
                "job_title"
            ],

        "required_experience":
            jd_data[
                "experience"
            ],

        "required_skills":
            normalized_jd_skills
    },

    "candidate_fit": {

        "overall_match_score":
            overall_score,

        "overall_status":
            overall_status
    },

    "resume": {

        "sections":
            sections
    },

    "skills": {

        "candidate_skills":
            sorted(
                candidate_skills,
                key=lambda x: x.lower()
            ),

        "matched_required_skills":
            matched_required_skills,

        "missing_required_skills":
            missing_required_skills,

        "required_skill_score":
            skill_score,

        "skill_evidence":
            skill_evidence
    },

    "experience": {

        "required_experience":
            f"{required_experience_years} year",

        "experience_duration":
            experience_duration,

        "experience_eligibility":
            experience_eligibility,

        "experience_evidence_score":
            experience_score,

        "evidence":
            experience_results
    },

    "projects": {

        "total_projects":
            len(project_results),

        "projects":
            project_results,

        "project_match_score":
            project_score
    },

    "score_breakdown": {

        "skills": {

            "score":
                skill_score,

            "weight":
                SKILL_WEIGHT,

            "contribution":
                round(
                    skill_score
                    * SKILL_WEIGHT,
                    2
                )
        },

        "experience": {

            "score":
                experience_score,

            "weight":
                EXPERIENCE_WEIGHT,

            "contribution":
                round(
                    experience_score
                    * EXPERIENCE_WEIGHT,
                    2
                )
        },

        "projects": {

            "score":
                project_score,

            "weight":
                PROJECT_WEIGHT,

            "contribution":
                round(
                    project_score
                    * PROJECT_WEIGHT,
                    2
                )
        }
    }
}


# ================================================================
# 42. HUMAN-READABLE REPORT
# ================================================================

print("\n\n")
print("=" * 80)
print("FINAL CANDIDATE EVALUATION")
print("=" * 80)


# ================================================================
# CANDIDATE
# ================================================================

print("\nCANDIDATE INFORMATION")
print("-" * 80)

print(
    "Name  :",
    candidate_name
)

print(
    "Phone :",
    candidate_phone
)

print(
    "Email :",
    candidate_email
)


# ================================================================
# JOB
# ================================================================

print("\nJOB INFORMATION")
print("-" * 80)

print(
    "Job Title:",
    jd_data[
        "job_title"
    ]
)

print(
    "Required Experience:",
    f"{required_experience_years} year"
)


# ================================================================
# SKILLS
# ================================================================

print("\nSKILL ANALYSIS")
print("-" * 80)

print(
    "Candidate Skills:"
)

for skill in sorted(
    candidate_skills,
    key=lambda x: x.lower()
):

    print(
        f"  • {skill}"
    )


print(
    "\nMatched Required Skills:"
)

for skill in matched_required_skills:

    print(
        f"  ✓ {skill}"
    )

    for item in skill_evidence[
        skill
    ]:

        print(
            f"      Source: "
            f"{item['source']}"
        )

        print(
            f"      Evidence: "
            f"{item['matched_text']}"
        )


print(
    "\nMissing Required Skills:"
)

for skill in missing_required_skills:

    print(
        f"  ✗ {skill}"
    )


print(
    f"\nSkill Match Score: "
    f"{skill_score:.2f}%"
)


# ================================================================
# EXPERIENCE
# ================================================================

print(
    "\nEXPERIENCE ANALYSIS"
)

print("-" * 80)


for requirement, result in (
    experience_results.items()
):

    print(
        f"\n{requirement}: "
        f"{result['strength']}"
    )

    if result["evidence"]:

        print(
            "   Evidence:",
            ", ".join(
                result["evidence"]
            )
        )

    else:

        print(
            "   Evidence: None"
        )


print(
    f"\nExperience Evidence Score: "
    f"{experience_score:.2f}%"
)

print(
    "Required Duration:",
    f"{required_experience_years} year"
)


if experience_duration[
    "total_years"
] is None:

    print(
        "Candidate Duration: Unknown"
    )

else:

    print(
        "Candidate Duration:",
        f"{experience_duration['total_years']} years"
    )

print(
    "Experience Eligibility:",
    experience_eligibility
)


# ================================================================
# PROJECTS
# ================================================================

print(
    "\nPROJECT ANALYSIS"
)

print("-" * 80)


if not project_results:

    print(
        "No projects detected."
    )

else:

    print(
        "Total Projects Detected:",
        len(project_results)
    )

    for index, project in enumerate(
        project_results,
        start=1
    ):

        print(
            f"\nProject {index}: "
            f"{project['title']}"
        )

        print(
            "Description:",
            project["description"]
        )

        print(
            "Project Match Score:",
            f"{project['project_match_score']:.2f}%"
        )

        for requirement, evidence in (
            project[
                "evidence"
            ].items()
        ):

            if evidence[
                "strength"
            ] != "NONE":

                print(
                    f"   {requirement}: "
                    f"{evidence['strength']}"
                )

                print(
                    "      Evidence:",
                    ", ".join(
                        evidence[
                            "evidence"
                        ]
                    )
                )


    print(
        f"\nOverall Project Score: "
        f"{project_score:.2f}%"
    )


# ================================================================
# SCORE BREAKDOWN
# ================================================================

print(
    "\nSCORE BREAKDOWN"
)

print("-" * 80)

print(
    f"Skills: "
    f"{skill_score:.2f}% × "
    f"{SKILL_WEIGHT} = "
    f"{skill_score * SKILL_WEIGHT:.2f}"
)

print(
    f"Experience: "
    f"{experience_score:.2f}% × "
    f"{EXPERIENCE_WEIGHT} = "
    f"{experience_score * EXPERIENCE_WEIGHT:.2f}"
)

print(
    f"Projects: "
    f"{project_score:.2f}% × "
    f"{PROJECT_WEIGHT} = "
    f"{project_score * PROJECT_WEIGHT:.2f}"
)


# ================================================================
# FINAL RESULT
# ================================================================

print(
    "\nFINAL RESULT"
)

print("-" * 80)

print(
    f"Overall Match Score: "
    f"{overall_score:.2f}%"
)

print(
    "Overall Status:",
    overall_status
)


# ================================================================
# 43. JSON OUTPUT
# ================================================================

print("\n\n")
print("=" * 80)
print("STRUCTURED JSON OUTPUT")
print("=" * 80)

print(
    json.dumps(
        evaluation_json,
        indent=4,
        ensure_ascii=False
    )
)


# ================================================================
# 44. OPTIONAL JSON SAVE
# ================================================================

# Uncomment this section if you want to save the result.
#
# OUTPUT_JSON_PATH = DATA_DIR / "evaluation_result.json"
#
# with open(
#     OUTPUT_JSON_PATH,
#     "w",
#     encoding="utf-8"
# ) as f:
#
#     json.dump(
#         evaluation_json,
#         f,
#         indent=4,
#         ensure_ascii=False
#     )
#
# print(
#     "\nJSON saved to:",
#     OUTPUT_JSON_PATH
# )
```

```python

# ================================================================
# AI-POWERED RESUME–JD MATCHING AND CANDIDATE EVALUATION SYSTEM
# ================================================================
#
# VERSION:
# Corrected baseline v2
#
# IMPORTANT:
# This is a RULE-BASED / HEURISTIC baseline.
# It is NOT yet a trained ML model.
#
# MAIN FIXES IN THIS VERSION:
# 1. PDF selectable-text extraction
# 2. OCR fallback
# 3. OCR cleanup
# 4. Resume section detection
# 5. Source-aware required-skill matching
# 6. Java != JavaScript
# 7. SQL != MySQL
# 8. React.js / ReactJS -> React
# 9. NodeJS / Node.js normalization
# 10. Role-aware experience detection
# 11. Technical and non-technical experience separation
# 12. Technical-only experience duration
# 13. Overlap-safe experience duration
# 14. Robust project title detection
# 15. "Tech:" / "Technology:" lines NEVER become projects
# 16. Technology lines are attached to the preceding project
# 17. Multi-line project grouping
# 18. Project scoring
# 19. Weighted final candidate score
# 20. Structured JSON output
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

from pathlib import Path
import re
import json
import io
from datetime import date

import fitz
import pytesseract
from PIL import Image


# ================================================================
# 2. PATH CONFIGURATION
# ================================================================

# This script is designed to run from:
# notebooks/InterviewQuestionGenerator.ipynb
#
# Therefore ".." points to the project root.
BASE_DIR = Path("..")
DATA_DIR = BASE_DIR / "data"

# ------------------------------------------------
# CHANGE ONLY THIS FILE NAME WHEN TESTING ANOTHER
# RESUME.
# ------------------------------------------------
RESUME_FILENAME = "1786378353416_PraptiPatil (1).pdf"

RESUME_PATH = DATA_DIR / RESUME_FILENAME

# JSON result will be saved here.
RESULT_JSON_PATH = DATA_DIR / "candidate_evaluation_result.json"


print("=" * 80)
print("RESUME–JD MATCHING SYSTEM")
print("=" * 80)

print("\nResume path:")
print(RESUME_PATH)

print("\nFile exists:", RESUME_PATH.exists())

if not RESUME_PATH.exists():
    raise FileNotFoundError(
        f"\nResume not found:\n{RESUME_PATH}\n"
        f"\nPlease check the filename in the data folder."
    )


# ================================================================
# 3. TESSERACT CONFIGURATION
# ================================================================

TESSERACT_PATH = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

pytesseract.pytesseract.tesseract_cmd = TESSERACT_PATH

try:
    print("\nTesseract version:", pytesseract.get_tesseract_version())
except Exception as exc:
    print("\nWarning: Tesseract could not be verified.")
    print("OCR may fail if Tesseract is not installed correctly.")
    print("Details:", exc)


# ================================================================
# 4. PDF TEXT EXTRACTION
# ================================================================

def extract_selectable_pdf_text(pdf_path):
    """Extract normal selectable text from every PDF page."""
    doc = fitz.open(pdf_path)

    page_texts = []

    for page in doc:
        page_texts.append(page.get_text())

    page_count = len(doc)
    doc.close()

    return "\n".join(page_texts), page_count


selectable_text, page_count = extract_selectable_pdf_text(RESUME_PATH)

print("\nSelectable PDF text characters:", len(selectable_text))
print("PDF pages:", page_count)


# ================================================================
# 5. OCR FALLBACK
# ================================================================

OCR_MIN_CHARACTERS = 100


def extract_ocr_text(pdf_path):
    """Render PDF pages and extract text using Tesseract OCR."""
    doc = fitz.open(pdf_path)

    ocr_pages = []

    for page_number, page in enumerate(doc, start=1):

        pix = page.get_pixmap(
            matrix=fitz.Matrix(2, 2),
            alpha=False
        )

        image = Image.open(
            io.BytesIO(
                pix.tobytes("png")
            )
        )

        page_text = pytesseract.image_to_string(image)

        ocr_pages.append(page_text)

        print(
            f"OCR completed: {page_number}/{len(doc)}"
        )

    doc.close()

    return "\n".join(ocr_pages)


if len(selectable_text.strip()) >= OCR_MIN_CHARACTERS:

    resume_text = selectable_text

    print("\nSelectable text detected.")
    print("OCR not required.")

else:

    print("\nSelectable text insufficient.")
    print("Running OCR...")

    resume_text = extract_ocr_text(RESUME_PATH)

    print("\nOCR extraction completed.")


# ================================================================
# 6. GENERAL TEXT CLEANUP
# ================================================================

def clean_resume_text(text):
    """Clean PDF/OCR text while preserving useful line structure."""

    text = text.replace("\r\n", "\n")
    text = text.replace("\r", "\n")

    # Join words broken by a line break:
    # deploy-
    # ed -> deployed
    text = re.sub(
        r"(\w)-\s*\n\s*(\w)",
        r"\1\2",
        text
    )

    # Normalize spaces without removing newlines.
    text = re.sub(
        r"[ \t]+",
        " ",
        text
    )

    # Remove excessive blank lines.
    text = re.sub(
        r"\n{3,}",
        "\n\n",
        text
    )

    return text.strip()


def controlled_ocr_normalization(text):
    """Correct only known OCR mistakes."""

    replacements = {
        "APls": "APIs",
        "APl": "API",
        "Al-driven": "AI-driven",
        "Al-powered": "AI-powered",
        "Al-driven": "AI-driven",
        "Gemini Al": "Gemini AI",
    }

    for old, new in replacements.items():
        text = text.replace(old, new)

    return text


resume_text = clean_resume_text(resume_text)
resume_text = controlled_ocr_normalization(resume_text)

print("\nFinal extracted characters:", len(resume_text))


# ================================================================
# 7. SECTION DEFINITIONS
# ================================================================

SECTION_ALIASES = {

    "Professional Summary": [
        "Professional Summary",
        "Summary",
        "Profile",
        "About Me",
        "Objective"
    ],

    "Skills": [
        "Technical Skills",
        "Technical Skill",
        "Skills",
        "Core Skills",
        "Key Skills",
        "Technologies",
        "Technical Expertise"
    ],

    "Internship & Experience": [
        "Professional Experience",
        "Experience",
        "Work Experience",
        "Internship",
        "Internships",
        "Internship & Experience",
        "Employment History",
        "Career Experience"
    ],

    "Projects": [
        "Projects",
        "Selected Projects",
        "Academic Projects",
        "Personal Projects",
        "Project Experience",
        "Projects & Research"
    ],

    "Education": [
        "Education",
        "Academic Background",
        "Educational Qualification",
        "Educational Qualifications"
    ],

    "Certifications": [
        "Certifications",
        "Certificates",
        "Professional Certifications"
    ],

    "Achievements": [
        "Achievements",
        "Awards",
        "Honors"
    ],

    "Leadership & Volunteering": [
        "Leadership & Volunteering",
        "Leadership",
        "Volunteering",
        "Positions of Responsibility"
    ]
}


def normalize_heading(line):
    """Normalize a possible section heading."""

    line = line.strip()

    line = re.sub(
        r"^[•●▪◦*\-]+\s*",
        "",
        line
    )

    line = re.sub(
        r"\s+",
        " ",
        line
    )

    return line.strip().lower()


HEADING_LOOKUP = {}

for section_name, aliases in SECTION_ALIASES.items():

    for alias in aliases:

        HEADING_LOOKUP[
            normalize_heading(alias)
        ] = section_name


# ================================================================
# 8. SECTION DETECTION
# ================================================================

def detect_sections(text):
    """Split the resume into named sections."""

    lines = text.splitlines()

    sections = {}

    current_section = None

    for raw_line in lines:

        line = raw_line.strip()

        if not line:
            continue

        normalized = normalize_heading(line)

        if normalized in HEADING_LOOKUP:

            current_section = HEADING_LOOKUP[normalized]

            if current_section not in sections:
                sections[current_section] = []

            continue

        if current_section is not None:
            sections[current_section].append(line)

    return sections


sections = detect_sections(resume_text)


print("\n" + "=" * 80)
print("DETECTED RESUME SECTIONS")
print("=" * 80)

for section_name, lines in sections.items():

    print(
        f"\n{section_name}: {len(lines)} lines"
    )


# ================================================================
# 9. BASIC CANDIDATE INFORMATION
# ================================================================

def extract_name(text):
    """Best-effort candidate name extraction."""

    lines = text.splitlines()

    for line in lines:

        line = line.strip()

        if not line:
            continue

        normalized = normalize_heading(line)

        if normalized in HEADING_LOOKUP:
            continue

        if "@" in line:
            continue

        if re.search(r"\d{7,}", line):
            continue

        if len(line.split()) > 5:
            continue

        if re.fullmatch(
            r"[A-Za-z][A-Za-z .'-]{1,60}",
            line
        ):
            return line

    return "Unknown"


def extract_phone(text):
    """Extract a likely phone number."""

    match = re.search(
        r"(?<!\d)(?:\+?\d[\d\s().-]{8,}\d)(?!\d)",
        text
    )

    if match:
        return match.group(0).strip()

    return "Unknown"


def extract_email(text):
    """Extract an email address."""

    match = re.search(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        text
    )

    if match:
        return match.group(0)

    return "Unknown"


candidate_name = extract_name(resume_text)
candidate_phone = extract_phone(resume_text)
candidate_email = extract_email(resume_text)


# ================================================================
# 10. JOB DESCRIPTION
# ================================================================

# This is the current JD being evaluated.
# Later this can be replaced with PDF/DOCX JD extraction.

jd_data = {

    "job_title":
        "Junior Full Stack Developer",

    "experience": {
        "min_years": 1,
        "max_years": None
    },

    "required_experience": [
        "full stack development",
        "web application development",
        "frontend development",
        "backend development",
        "REST API development",
        "database development"
    ],

    "required_skills": [
        "Java",
        "Spring Boot",
        "MySQL",
        "HTML",
        "CSS",
        "JavaScript",
        "React",
        "Node.js"
    ],

    "preferred_skills": []
}


# ================================================================
# 11. SKILL NORMALIZATION
# ================================================================

def normalize_skill(skill):
    """Normalize common technology aliases."""

    skill = skill.strip()

    skill = re.sub(
        r"^[•●▪◦*\-]+\s*",
        "",
        skill
    )

    skill = re.sub(
        r"\s+",
        " ",
        skill
    )

    skill = skill.strip(" ,;:")

    normalized = skill.lower()

    aliases = {

        "react.js": "React",
        "reactjs": "React",

        "nodejs": "Node.js",
        "node.js": "Node.js",

        "html5": "HTML",
        "html": "HTML",

        "css3": "CSS",
        "css": "CSS",

        "springboot": "Spring Boot",
        "spring boot": "Spring Boot",

        "mysql": "MySQL",

        "mongodb": "MongoDB",

        "sql": "SQL",

        "express": "Express.js",
        "expressjs": "Express.js",
        "express.js": "Express.js",

        "c++": "C++",

        "github": "GitHub",
        "gitlab": "GitLab"
    }

    return aliases.get(
        normalized,
        skill
    )


def split_top_level_commas(text):
    """Split commas while ignoring commas inside parentheses."""

    parts = []
    current = []
    depth = 0

    for char in text:

        if char == "(":
            depth += 1

        elif char == ")":
            depth = max(0, depth - 1)

        if char == "," and depth == 0:

            part = "".join(current).strip()

            if part:
                parts.append(part)

            current = []

        else:
            current.append(char)

    final_part = "".join(current).strip()

    if final_part:
        parts.append(final_part)

    return parts


def extract_candidate_skills(skills_lines):
    """Extract skills from the Skills section."""

    extracted = []

    for line in skills_lines:

        line = line.strip()

        if not line:
            continue

        line = re.sub(
            r"^[•●▪◦*\-]+\s*",
            "",
            line
        )

        # Remove category such as:
        # Programming: Java, Python
        if ":" in line:
            _, values = line.split(":", 1)
            values = values.strip()
        else:
            values = line

        items = split_top_level_commas(values)

        for item in items:

            item = item.strip()

            if not item:
                continue

            # Handle:
            # AWS (Lambda, Amplify, S3)
            parenthetical = re.fullmatch(
                r"(.+?)\s*\(([^()]*)\)",
                item
            )

            if parenthetical:

                main_skill = parenthetical.group(1).strip()

                normalized_main = normalize_skill(
                    main_skill
                )

                if normalized_main:
                    extracted.append(normalized_main)

                for sub_skill in split_top_level_commas(
                    parenthetical.group(2)
                ):

                    normalized_sub = normalize_skill(
                        sub_skill
                    )

                    if normalized_sub:
                        extracted.append(normalized_sub)

            else:

                normalized = normalize_skill(item)

                if normalized:
                    extracted.append(normalized)

    # Remove duplicates.
    unique = []
    seen = set()

    for skill in extracted:

        key = skill.lower()

        if key not in seen:

            seen.add(key)
            unique.append(skill)

    return unique


candidate_skills = extract_candidate_skills(
    sections.get("Skills", [])
)


# ================================================================
# 12. REQUIRED JD SKILL REGEX
# ================================================================

normalized_jd_skills = [
    normalize_skill(skill)
    for skill in jd_data["required_skills"]
]


SKILL_PATTERNS = {

    # IMPORTANT:
    # Java must NOT match JavaScript.
    "Java": [
        r"(?<!\w)java(?!script)(?!\w)"
    ],

    "Spring Boot": [
        r"(?<!\w)spring\s*boot(?!\w)",
        r"(?<!\w)springboot(?!\w)"
    ],

    "MySQL": [
        r"(?<!\w)mysql(?!\w)"
    ],

    "HTML": [
        r"(?<!\w)html(?:5)?(?!\w)"
    ],

    "CSS": [
        r"(?<!\w)css(?:3)?(?!\w)"
    ],

    "JavaScript": [
        r"(?<!\w)javascript(?!\w)"
    ],

    "React": [
        r"(?<!\w)react(?:\.js|js)?(?!\w)"
    ],

    "Node.js": [
        r"(?<!\w)node(?:\.js|js)(?!\w)"
    ]
}


def find_skill_evidence(skill, text):
    """Find evidence for a required skill."""

    patterns = SKILL_PATTERNS.get(
        skill,
        [
            r"(?<!\w)"
            + re.escape(skill)
            + r"(?!\w)"
        ]
    )

    for pattern in patterns:

        match = re.search(
            pattern,
            text,
            re.IGNORECASE
        )

        if match:

            start = max(
                0,
                match.start() - 70
            )

            end = min(
                len(text),
                match.end() + 70
            )

            return {
                "matched": True,
                "matched_text": match.group(0),
                "context": text[start:end].strip()
            }

    return {
        "matched": False,
        "matched_text": None,
        "context": None
    }


# ================================================================
# 13. SOURCE-AWARE REQUIRED SKILL MATCHING
# ================================================================

SOURCE_PRIORITY = [
    "Skills",
    "Internship & Experience",
    "Projects",
    "Professional Summary"
]


def match_required_skills_with_sources(jd_skills, sections):

    matched = []
    missing = []
    evidence = {}

    for required_skill in jd_skills:

        skill_sources = []

        for source in SOURCE_PRIORITY:

            source_lines = sections.get(
                source,
                []
            )

            source_text = "\n".join(source_lines)

            result = find_skill_evidence(
                required_skill,
                source_text
            )

            if result["matched"]:

                skill_sources.append({
                    "source": source,
                    "matched_text":
                        result["matched_text"],
                    "context":
                        result["context"]
                })

        if skill_sources:

            matched.append(required_skill)

            evidence[required_skill] = skill_sources

        else:

            missing.append(required_skill)
            evidence[required_skill] = []

    score = (
        len(matched)
        / len(jd_skills)
        * 100
        if jd_skills
        else 0
    )

    return (
        matched,
        missing,
        evidence,
        round(score, 2)
    )


(
    matched_required_skills,
    missing_required_skills,
    skill_evidence,
    skill_score
) = match_required_skills_with_sources(
    normalized_jd_skills,
    sections
)


# ================================================================
# 14. EXPERIENCE RULES
# ================================================================

experience_evidence_rules = {

    "full stack development": [
        r"\bfull[- ]stack\b",
        r"\bend[- ]to[- ]end\b",
        r"\bmern\b",
        r"\bmean\b",
        r"\bfrontend\s+and\s+backend\b"
    ],

    "web application development": [
        r"\bweb[- ]based\s+applications?\b",
        r"\bweb\s+applications?\b",
        r"\bweb\s+platform\b",
        r"\bdynamic\s+web\b",
        r"\bcontent\s+management\s+panel\b"
    ],

    "frontend development": [
        r"\bfrontend\b",
        r"\bfront[- ]end\b",
        r"\buser\s+interface\b",
        r"\bUI\b"
    ],

    "backend development": [
        r"\bbackend\b",
        r"\bback[- ]end\b",
        r"\bserver[- ]side\b",
        r"\bserver\s+side\b"
    ],

    "REST API development": [
        r"\bRESTful?\s+API(?:s)?\b",
        r"\bREST\s+API(?:s)?\b",
        r"\bAPI\s+development\b",
        r"\bcustom\s+.*?\bAPI(?:s)?\b",
        r"\bAPI(?:s)?\b"
    ],

    "database development": [
        r"\bdatabase\b",
        r"\bdatabases\b",
        r"\bdatabase\s+queries\b",
        r"\brelational\s+database\b",
        r"\bMongoDB\b",
        r"\bMySQL\b",
        r"\bSQL\b",
        r"\bFirebase\b"
    ]
}


experience_strong_patterns = {

    "full stack development": [
        r"\bfull[- ]stack\b",
        r"\bend[- ]to[- ]end\b",
        r"\bmern\b",
        r"\bmean\b"
    ],

    "web application development": [
        r"\bweb[- ]based\s+applications?\b",
        r"\bweb\s+applications?\b",
        r"\bweb\s+platform\b",
        r"\bcontent\s+management\s+panel\b"
    ],

    "frontend development": [
        r"\bfrontend\b",
        r"\bfront[- ]end\b",
        r"\buser\s+interface\b",
        r"\bUI\b"
    ],

    "backend development": [
        r"\bbackend\b",
        r"\bback[- ]end\b",
        r"\bserver[- ]side\b",
        r"\bserver\s+side\b"
    ],

    "REST API development": [
        r"\bRESTful?\s+API(?:s)?\b",
        r"\bREST\s+API(?:s)?\b",
        r"\bAPI\s+development\b",
        r"\bcustom\s+.*?\bAPI(?:s)?\b"
    ],

    "database development": [
        r"\bdatabase\b",
        r"\bdatabases\b",
        r"\bdatabase\s+queries\b",
        r"\brelational\s+database\b"
    ]
}


# ================================================================
# 15. EXPERIENCE ROLE CLASSIFICATION
# ================================================================

TECHNICAL_ROLE_KEYWORDS = [
    "developer",
    "engineer",
    "software",
    "programmer",
    "full stack",
    "frontend",
    "front-end",
    "backend",
    "back-end",
    "web developer",
    "software developer",
    "software engineer",
    "developer intern",
    "technical",
    "data analyst",
    "data engineer",
    "data scientist",
    "ai developer",
    "ml engineer",
    "machine learning",
    "devops",
    "cloud engineer",
    "qa engineer",
    "automation tester",
    "software intern"
]


NON_TECHNICAL_ROLE_KEYWORDS = [
    "event",
    "co-head",
    "co head",
    "president",
    "vice president",
    "coordinator",
    "volunteer",
    "marketing",
    "sales",
    "human resources",
    "hr",
    "recruiter",
    "finance",
    "accountant",
    "content writer",
    "social media",
    "public relations",
    "operations"
]


def is_bullet_line(line):
    return bool(
        re.match(
            r"^\s*[•●▪◦*]\s*",
            line
        )
    )


def clean_line(line):
    line = line.strip()

    line = re.sub(
        r"^\s*[•●▪◦*]+\s*",
        "",
        line
    )

    return line.strip()


def classify_role(role_title, company, body):

    role_text = clean_line(role_title).lower()
    company_text = clean_line(company).lower()
    body_text = body.lower()

    # Role title has the highest priority.
    if any(
        keyword in role_text
        for keyword in NON_TECHNICAL_ROLE_KEYWORDS
    ):

        if not any(
            keyword in role_text
            for keyword in TECHNICAL_ROLE_KEYWORDS
        ):

            return "NON_TECHNICAL"

    if any(
        keyword in role_text
        for keyword in TECHNICAL_ROLE_KEYWORDS
    ):
        return "TECHNICAL"

    # If title is ambiguous, inspect the company + body,
    # but require multiple technical signals.
    strong_body_signals = [
        r"\bfull[- ]stack\b",
        r"\bfrontend\b",
        r"\bfront[- ]end\b",
        r"\bbackend\b",
        r"\bback[- ]end\b",
        r"\bnode(?:\.js|js)\b",
        r"\breact(?:\.js|js)?\b",
        r"\bpython\b",
        r"\bjava\b",
        r"\bAPI\b",
        r"\bdatabase\b",
        r"\bsoftware\b",
        r"\bweb application\b",
        r"\bserver[- ]side\b"
    ]

    body_signal_count = sum(
        1
        for pattern in strong_body_signals
        if re.search(
            pattern,
            body_text,
            re.IGNORECASE
        )
    )

    company_signal_count = sum(
        1
        for keyword in TECHNICAL_ROLE_KEYWORDS
        if keyword in company_text
    )

    if body_signal_count >= 2 or company_signal_count >= 1:
        return "TECHNICAL"

    return "NON_TECHNICAL"


# ================================================================
# 16. DATE PARSING
# ================================================================

MONTH_MAP = {

    "jan": 1,
    "january": 1,

    "feb": 2,
    "february": 2,

    "mar": 3,
    "march": 3,

    "apr": 4,
    "april": 4,

    "may": 5,

    "jun": 6,
    "june": 6,

    "jul": 7,
    "july": 7,

    "aug": 8,
    "august": 8,

    "sep": 9,
    "sept": 9,
    "september": 9,

    "oct": 10,
    "october": 10,

    "nov": 11,
    "november": 11,

    "dec": 12,
    "december": 12
}


MONTH_PATTERN = (
    r"(?:Jan(?:uary)?|"
    r"Feb(?:ruary)?|"
    r"Mar(?:ch)?|"
    r"Apr(?:il)?|"
    r"May|"
    r"Jun(?:e)?|"
    r"Jul(?:y)?|"
    r"Aug(?:ust)?|"
    r"Sep(?:t(?:ember)?)?|"
    r"Oct(?:ober)?|"
    r"Nov(?:ember)?|"
    r"Dec(?:ember)?)"
)


DATE_RANGE_PATTERN = re.compile(
    rf"(?P<start>{MONTH_PATTERN}\s+\d{{4}})"
    rf"\s*(?:–|—|-|to)\s*"
    rf"(?P<end>{MONTH_PATTERN}\s+\d{{4}}|Present|Current)",
    re.IGNORECASE
)


def parse_month_year(text):
    """Convert Month Year into a date object."""

    text = text.strip().lower()

    match = re.search(
        rf"\b({MONTH_PATTERN})\s+(\d{{4}})",
        text,
        re.IGNORECASE
    )

    if not match:
        return None

    month_text = match.group(1).lower()
    year = int(match.group(2))

    month = MONTH_MAP.get(month_text)

    if not month:
        return None

    return date(
        year,
        month,
        1
    )


def parse_date_range_from_line(line):
    """Return a date-range match from a single resume line."""

    match = DATE_RANGE_PATTERN.search(line)

    if not match:
        return None

    start_date = parse_month_year(
        match.group("start")
    )

    end_text = match.group("end").strip()

    if end_text.lower() in {"present", "current"}:
        end_date = date.today()
        end_label = end_text
        is_current = True
    else:
        end_date = parse_month_year(end_text)
        end_label = end_text
        is_current = False

    if not start_date or not end_date:
        return None

    return {
        "start": match.group("start"),
        "end": end_label,
        "start_date": start_date,
        "end_date": end_date,
        "is_current": is_current,
        "span_start": match.start(),
        "span_end": match.end()
    }


# ================================================================
# 17. ROBUST EXPERIENCE BLOCK EXTRACTION
# ================================================================

def find_closest_role_title(lines, date_index):

    # ------------------------------------------------------------
    # 1. If the role title and date are on the SAME line,
    #    use the text before the date.
    # ------------------------------------------------------------
    current_line = clean_line(lines[date_index])

    date_info = parse_date_range_from_line(current_line)

    if date_info:

        inline_role = (
            current_line[:date_info["span_start"]]
            .strip(" |:-–—")
            .strip()
        )

        if inline_role:
            return inline_role

    # ------------------------------------------------------------
    # 2. If the date is on a separate line, inspect the previous
    #    few lines. Do NOT discard bullets: resume role titles are
    #    very often bullet-prefixed.
    # ------------------------------------------------------------
    candidates = []

    start_index = max(0, date_index - 5)

    for index in range(
        date_index - 1,
        start_index - 1,
        -1
    ):

        raw_candidate = lines[index].strip()

        if not raw_candidate:
            continue

        candidate = clean_line(raw_candidate)

        if not candidate:
            continue

        if parse_date_range_from_line(candidate):
            continue

        if len(candidate) > 120:
            continue

        candidates.append(candidate)

    # ------------------------------------------------------------
    # 3. Strongly prefer an explicit role title.
    #    This fixes cases such as:
    #
    #    Event Co-Head – Phoenix Club, KKWIEER
    #    Nov 2025 - June 2026
    #
    #    where a preceding descriptive sentence exists.
    # ------------------------------------------------------------
    for candidate in candidates:

        lower_candidate = candidate.lower()

        if any(
            keyword in lower_candidate
            for keyword in NON_TECHNICAL_ROLE_KEYWORDS
        ):
            return candidate

    for candidate in candidates:

        lower_candidate = candidate.lower()

        if any(
            keyword in lower_candidate
            for keyword in TECHNICAL_ROLE_KEYWORDS
        ):
            return candidate

    # ------------------------------------------------------------
    # 4. Prefer short title-like candidates over long sentences.
    # ------------------------------------------------------------
    short_candidates = [
        candidate
        for candidate in candidates
        if len(candidate.split()) <= 12
    ]

    if short_candidates:
        return short_candidates[0]

    if candidates:
        return candidates[0]

    return "Unknown Role"


def extract_experience_blocks(experience_lines):

    blocks = []

    date_indexes = []

    for index, line in enumerate(experience_lines):

        if parse_date_range_from_line(line):
            date_indexes.append(index)

    for position, date_index in enumerate(date_indexes):

        date_info = parse_date_range_from_line(
            experience_lines[date_index]
        )

        next_date_index = (
            date_indexes[position + 1]
            if position + 1 < len(date_indexes)
            else len(experience_lines)
        )

        role_title = find_closest_role_title(
            experience_lines,
            date_index
        )

        # Company may appear immediately after the date line.
        company = "Unknown Company"

        after_date_candidates = []

        for index in range(
            date_index + 1,
            min(
                next_date_index,
                date_index + 3
            )
        ):

            candidate = clean_line(
                experience_lines[index]
            )

            if not candidate:
                continue

            if parse_date_range_from_line(candidate):
                continue

            if is_bullet_line(
                experience_lines[index]
            ):
                continue

            if len(candidate) <= 100:
                after_date_candidates.append(candidate)

        if after_date_candidates:
            company = after_date_candidates[0]

        # Body starts after the date line.
        body_lines = []

        for index in range(
            date_index + 1,
            next_date_index
        ):

            line = clean_line(
                experience_lines[index]
            )

            if not line:
                continue

            # Do not duplicate a company line as body.
            if (
                line == company
                and
                not is_bullet_line(
                    experience_lines[index]
                )
            ):
                continue

            body_lines.append(line)

        body = " ".join(body_lines).strip()

        role_type = classify_role(
            role_title,
            company,
            body
        )

        blocks.append({
            "role_title": role_title,
            "company": company,
            "date_text": (
                experience_lines[date_index].strip()
            ),
            "start": date_info["start"],
            "end": date_info["end"],
            "start_date": date_info["start_date"],
            "end_date": date_info["end_date"],
            "is_current": date_info["is_current"],
            "role_type": role_type,
            "body": body
        })

    return blocks


experience_lines = sections.get(
    "Internship & Experience",
    []
)

experience_blocks = extract_experience_blocks(
    experience_lines
)


# ================================================================
# 18. TECHNICAL-ONLY EXPERIENCE DURATION
# ================================================================

def calculate_inclusive_months(
    start_date,
    end_date
):

    if end_date < start_date:
        return 0

    return (
        (end_date.year - start_date.year) * 12
        + (end_date.month - start_date.month)
        + 1
    )


def merge_date_intervals(intervals):

    if not intervals:
        return []

    intervals = sorted(
        intervals,
        key=lambda item: item[0]
    )

    merged = [list(intervals[0])]

    for current_start, current_end in intervals[1:]:

        previous_start, previous_end = merged[-1]

        # Convert dates into month indexes.
        previous_end_index = (
            previous_end.year * 12
            + previous_end.month
        )

        current_start_index = (
            current_start.year * 12
            + current_start.month
        )

        if current_start_index <= previous_end_index + 1:

            if current_end > previous_end:
                merged[-1][1] = current_end

        else:

            merged.append([
                current_start,
                current_end
            ])

    return [
        (start, end)
        for start, end in merged
    ]


def calculate_technical_experience(blocks):

    technical_blocks = [
        block
        for block in blocks
        if block["role_type"] == "TECHNICAL"
    ]

    non_technical_blocks = [
        block
        for block in blocks
        if block["role_type"] == "NON_TECHNICAL"
    ]

    intervals = [
        (
            block["start_date"],
            block["end_date"]
        )
        for block in technical_blocks
    ]

    merged_intervals = merge_date_intervals(
        intervals
    )

    total_months = sum(
        calculate_inclusive_months(
            start,
            end
        )
        for start, end in merged_intervals
    )

    if technical_blocks:

        return {
            "status":
                "CALCULATED_FROM_TECHNICAL_ROLES",

            "total_months":
                total_months,

            "total_years":
                round(
                    total_months / 12,
                    2
                ),

            "technical_roles":
                [
                    {
                        "role":
                            block["role_title"],
                        "company":
                            block["company"],
                        "start":
                            block["start"],
                        "end":
                            block["end"],
                        "months":
                            calculate_inclusive_months(
                                block["start_date"],
                                block["end_date"]
                            )
                    }
                    for block in technical_blocks
                ],

            "excluded_non_technical_roles":
                [
                    {
                        "role":
                            block["role_title"],
                        "company":
                            block["company"],
                        "start":
                            block["start"],
                        "end":
                            block["end"]
                    }
                    for block in non_technical_blocks
                ],

            "merged_intervals":
                [
                    {
                        "start":
                            start.strftime("%b %Y"),
                        "end":
                            end.strftime("%b %Y"),
                        "months":
                            calculate_inclusive_months(
                                start,
                                end
                            )
                    }
                    for start, end in merged_intervals
                ]
        }

    return {
        "status": "UNKNOWN",
        "total_months": None,
        "total_years": None,
        "technical_roles": [],
        "excluded_non_technical_roles": [
            {
                "role": block["role_title"],
                "company": block["company"],
                "start": block["start"],
                "end": block["end"]
            }
            for block in non_technical_blocks
        ],
        "merged_intervals": []
    }


experience_duration = calculate_technical_experience(
    experience_blocks
)


required_experience_years = jd_data[
    "experience"
]["min_years"]


if experience_duration["total_years"] is None:

    experience_eligibility = "UNKNOWN"

elif (
    experience_duration["total_years"]
    >= required_experience_years
):

    experience_eligibility = "MEETS_REQUIREMENT"

else:

    experience_eligibility = "BELOW_REQUIREMENT"


# ================================================================
# 19. EXPERIENCE EVIDENCE
# ================================================================

technical_experience_text = "\n".join(
    (
        block["role_title"]
        + " "
        + block["company"]
        + " "
        + block["body"]
    )
    for block in experience_blocks
    if block["role_type"] == "TECHNICAL"
)


def evaluate_experience_evidence(
    experience_text
):

    results = {}
    total_score = 0

    for requirement, patterns in (
        experience_evidence_rules.items()
    ):

        found = []

        for pattern in patterns:

            for match in re.finditer(
                pattern,
                experience_text,
                re.IGNORECASE
            ):

                matched_text = match.group(0).strip()

                if matched_text not in found:
                    found.append(matched_text)

        strong_patterns = experience_strong_patterns.get(
            requirement,
            []
        )

        strong_found = []

        for pattern in strong_patterns:

            match = re.search(
                pattern,
                experience_text,
                re.IGNORECASE
            )

            if match:
                strong_found.append(
                    match.group(0)
                )

        if strong_found:

            strength = "STRONG"
            numeric_score = 1.0

        elif found:

            strength = "PARTIAL"
            numeric_score = 0.5

        else:

            strength = "NONE"
            numeric_score = 0.0

        total_score += numeric_score

        results[requirement] = {
            "evidence": found,
            "strength": strength,
            "numeric_score": numeric_score
        }

    overall_score = (
        total_score
        / len(experience_evidence_rules)
        * 100
        if experience_evidence_rules
        else 0
    )

    return (
        results,
        round(overall_score, 2)
    )


experience_results, experience_score = (
    evaluate_experience_evidence(
        technical_experience_text
    )
)


# ================================================================
# 20. PROJECT DETECTION HELPERS
# ================================================================

PROJECT_METADATA_PREFIXES = (
    "tech:",
    "tech stack:",
    "technologies:",
    "technology:",
    "technologies used:",
    "technology used:",
    "tools:",
    "tools used:",
    "built with:",
    "stack:",
    "tools/technologies:",
    "tools / technologies:"
)


PROJECT_DESCRIPTION_STARTS = (
    "built ",
    "developed ",
    "designed ",
    "created ",
    "implemented ",
    "deployed ",
    "integrated ",
    "used ",
    "leveraged ",
    "improved ",
    "worked ",
    "participated ",
    "an ai ",
    "a web ",
    "a mobile ",
    "this ",
    "the system ",
    "the platform "
)


def is_project_metadata_line(line):

    cleaned = clean_line(line).lower()

    return cleaned.startswith(
        PROJECT_METADATA_PREFIXES
    )


def split_project_title_and_description(line):

    cleaned = clean_line(line)

    if not cleaned:
        return None, None

    # Metadata is NEVER a project title.
    if is_project_metadata_line(cleaned):
        return None, None

    # Support:
    # AICoRex – Description
    # AICoRex — Description
    dash_match = re.split(
        r"\s+[–—]\s+|\s+[–—]\s*",
        cleaned,
        maxsplit=1
    )

    if len(dash_match) == 2:

        title = dash_match[0].strip()
        description = dash_match[1].strip()

        if (
            title
            and len(title) <= 100
            and not title.lower().startswith(
                ("tech", "technology", "tools")
            )
        ):
            return title, description

    # Support:
    # Project Name - Description
    hyphen_match = re.split(
        r"\s+-\s+",
        cleaned,
        maxsplit=1
    )

    if len(hyphen_match) == 2:

        title = hyphen_match[0].strip()
        description = hyphen_match[1].strip()

        # Do not split normal descriptive sentences.
        if (
            title
            and description
            and len(title) <= 100
            and not title.lower().startswith(
                ("tech", "technology", "tools")
            )
        ):

            first_word = title.split()[0].lower()

            if first_word not in {
                "built",
                "developed",
                "designed",
                "created",
                "implemented"
            }:
                return title, description

    return None, None


def looks_like_standalone_project_title(line):

    cleaned = clean_line(line)

    if not cleaned:
        return False

    lower_line = cleaned.lower()

    # ------------------------------------------------------------
    # CRITICAL FIX:
    # Never treat technology metadata as a project title.
    # ------------------------------------------------------------
    if is_project_metadata_line(cleaned):
        return False

    # Never treat obvious descriptions as titles.
    if lower_line.startswith(
        PROJECT_DESCRIPTION_STARTS
    ):
        return False

    # Ignore long sentences.
    if len(cleaned) > 100:
        return False

    # Ignore lines containing obvious technology metadata.
    if re.match(
        r"^(tech|technology|technologies|tools|stack)\s*:",
        lower_line
    ):
        return False

    # Ignore sentences ending in punctuation.
    if cleaned.endswith(
        (".", "!", "?")
    ):
        return False

    # Project-like keywords.
    project_keywords = (
        "system",
        "platform",
        "application",
        "app",
        "engine",
        "portal",
        "website",
        "management",
        "planner"
    )

    if any(
        re.search(
            rf"\b{re.escape(keyword)}\b",
            lower_line
        )
        for keyword in project_keywords
    ):
        return True

    # ------------------------------------------------------------
    # Single-line project names such as:
    # AICoRex
    # DietGenius
    # VIKRSHI
    #
    # These may not contain "system/platform/app".
    # ------------------------------------------------------------
    words = re.findall(
        r"[A-Za-z][A-Za-z0-9&.'+-]*",
        cleaned
    )

    if not words:
        return False

    # Very short 1–4 word title-like lines are accepted
    # if they contain no obvious description verbs.
    if len(words) <= 4:

        verb_pattern = (
            r"\b(is|are|was|were|has|have|uses|used|"
            r"built|developed|designed|created|implemented|"
            r"provides|providing|helps|helping)\b"
        )

        if not re.search(
            verb_pattern,
            lower_line
        ):
            return True

    # Title-case heuristic for multi-word names.
    capitalized = sum(
        1
        for word in words
        if word[0].isupper()
    )

    ratio = capitalized / len(words)

    return ratio >= 0.60


def finalize_project(project):

    if project is None:
        return None

    description_lines = project.get(
        "description_lines",
        []
    )

    description = " ".join(
        line.strip()
        for line in description_lines
        if line.strip()
    ).strip()

    return {
        "title": project["title"].strip(),
        "description": description,
        "description_lines": description_lines
    }


# ================================================================
# 21. ROBUST PROJECT EXTRACTION
# ================================================================

def extract_projects(project_lines):

    projects = []

    current_project = None

    for raw_line in project_lines:

        line = raw_line.strip()

        if not line:
            continue

        cleaned = clean_line(line)

        if not cleaned:
            continue

        # --------------------------------------------------------
        # 1. Technology metadata ALWAYS belongs to current project.
        # --------------------------------------------------------
        if is_project_metadata_line(cleaned):

            if current_project is not None:
                current_project[
                    "description_lines"
                ].append(cleaned)

            # If there is no current project, keep it pending.
            # It must NOT create a project.
            continue

        # --------------------------------------------------------
        # 2. Try same-line:
        #
        # Project – Description
        # --------------------------------------------------------
        title, description = (
            split_project_title_and_description(
                line
            )
        )

        if title:

            finished = finalize_project(
                current_project
            )

            if finished:
                projects.append(finished)

            current_project = {
                "title": title,
                "description_lines": []
            }

            if description:
                current_project[
                    "description_lines"
                ].append(description)

            continue

        # --------------------------------------------------------
        # 3. Standalone project title.
        #
        # Important:
        # Bullet status does NOT prevent title detection.
        # A resume may write:
        # • AICoRex
        # • DietGenius
        # --------------------------------------------------------
        if looks_like_standalone_project_title(
            line
        ):

            finished = finalize_project(
                current_project
            )

            if finished:
                projects.append(finished)

            current_project = {
                "title": cleaned,
                "description_lines": []
            }

            continue

        # --------------------------------------------------------
        # 4. Any remaining line belongs to current project.
        # --------------------------------------------------------
        if current_project is not None:

            current_project[
                "description_lines"
            ].append(cleaned)

    # Save final project.
    finished = finalize_project(
        current_project
    )

    if finished:
        projects.append(finished)

    # Remove accidental duplicates while preserving order.
    unique_projects = []
    seen_titles = set()

    for project in projects:

        key = re.sub(
            r"\s+",
            " ",
            project["title"].lower()
        ).strip()

        if key in seen_titles:
            continue

        seen_titles.add(key)
        unique_projects.append(project)

    return unique_projects


projects = extract_projects(
    sections.get(
        "Projects",
        []
    )
)


# ================================================================
# 22. PROJECT EVIDENCE RULES
# ================================================================

project_evidence_rules = {

    "full stack development": [
        r"\bfull[- ]stack\b",
        r"\bmern\b",
        r"\bmean\b",
        r"\bend[- ]to[- ]end\b",
        r"\bfrontend\s+and\s+backend\b"
    ],

    "web application development": [
        r"\bweb\s+application\b",
        r"\bweb\s+applications\b",
        r"\bweb[- ]based\s+application\b",
        r"\bweb[- ]based\s+applications\b",
        r"\bweb\s+platform\b",
        r"\bweb[- ]platform\b",
        r"\bweb\s+portal\b",
        r"\bwebsite\b",
        r"\bdynamic\s+web\b"
    ],

    "frontend development": [
        r"\breact(?:\.js|js)?\b",
        r"\breact[- ]native\b",
        r"\breact-based\b",
        r"\bfrontend\b",
        r"\bfront[- ]end\b",
        r"\buser\s+interface\b",
        r"\bUI\b"
    ],

    "backend development": [
        r"\bbackend\b",
        r"\bback[- ]end\b",
        r"\bserver[- ]side\b",
        r"\bserver\s+side\b",
        r"\bnode(?:\.js|js)\b",
        r"\bPHP\b",
        r"\bpython\b"
    ],

    "REST API development": [
        r"\bRESTful?\s+API(?:s)?\b",
        r"\bREST\s+API(?:s)?\b",
        r"\bcustom\s+.*?\bAPI(?:s)?\b",
        r"\bAPI\s+development\b",
        r"\bAPI\s+integration\b",
        r"\bAPI\s+integrations\b",
        r"\bAPI\b"
    ],

    "database development": [
        r"\bdatabase\b",
        r"\bdatabases\b",
        r"\bMongoDB\b",
        r"\bMySQL\b",
        r"\bSQL\b",
        r"\bFirebase\b",
        r"\bFirestore\b",
        r"\bPostgreSQL\b",
        r"\bConvex\b"
    ]
}


# ================================================================
# 23. PROJECT EVALUATION
# ================================================================

def evaluate_project(project):

    title = project["title"]
    description = project["description"]

    searchable_text = (
        title
        + " "
        + description
    )

    evidence_result = {}

    total_score = 0

    for requirement, patterns in (
        project_evidence_rules.items()
    ):

        found = []

        for pattern in patterns:

            for match in re.finditer(
                pattern,
                searchable_text,
                re.IGNORECASE
            ):

                matched = match.group(0).strip()

                if matched not in found:
                    found.append(matched)

        # --------------------------------------------------------
        # Full stack requires explicit evidence.
        # --------------------------------------------------------
        if requirement == "full stack development":

            strong_patterns = [
                r"\bfull[- ]stack\b",
                r"\bmern\b",
                r"\bmean\b",
                r"\bend[- ]to[- ]end\b"
            ]

            strong = any(
                re.search(
                    pattern,
                    searchable_text,
                    re.IGNORECASE
                )
                for pattern in strong_patterns
            )

            if strong:

                strength = "STRONG"
                numeric_score = 1.0

            else:

                strength = "NONE"
                numeric_score = 0.0
                found = []

        # --------------------------------------------------------
        # REST API:
        # REST API / custom API = strong
        # Generic API = partial
        # --------------------------------------------------------
        elif requirement == "REST API development":

            strong_patterns = [
                r"\bRESTful?\s+API(?:s)?\b",
                r"\bREST\s+API(?:s)?\b",
                r"\bcustom\s+.*?\bAPI(?:s)?\b",
                r"\bAPI\s+development\b"
            ]

            strong = any(
                re.search(
                    pattern,
                    searchable_text,
                    re.IGNORECASE
                )
                for pattern in strong_patterns
            )

            if strong:

                strength = "STRONG"
                numeric_score = 1.0

            elif found:

                strength = "PARTIAL"
                numeric_score = 0.5

            else:

                strength = "NONE"
                numeric_score = 0.0

        else:

            if found:

                strength = "STRONG"
                numeric_score = 1.0

            else:

                strength = "NONE"
                numeric_score = 0.0

        total_score += numeric_score

        evidence_result[requirement] = {
            "evidence": found,
            "strength": strength,
            "numeric_score": numeric_score
        }

    project_score = (
        total_score
        / len(project_evidence_rules)
        * 100
        if project_evidence_rules
        else 0
    )

    return {
        "title": title,
        "description": description,
        "evidence": evidence_result,
        "project_match_score":
            round(project_score, 2)
    }


project_results = [
    evaluate_project(project)
    for project in projects
]


# Average all detected projects.
if project_results:

    project_score = round(
        sum(
            project["project_match_score"]
            for project in project_results
        )
        / len(project_results),
        2
    )

else:

    project_score = 0.0


# ================================================================
# 24. OVERALL SCORE
# ================================================================

SKILL_WEIGHT = 0.50
EXPERIENCE_WEIGHT = 0.30
PROJECT_WEIGHT = 0.20

overall_score = (
    skill_score * SKILL_WEIGHT
    + experience_score * EXPERIENCE_WEIGHT
    + project_score * PROJECT_WEIGHT
)

overall_score = round(
    overall_score,
    2
)


# ================================================================
# 25. STATUS
# ================================================================

def score_status(score):

    if score >= 70:
        return "HIGH MATCH"

    elif score >= 40:
        return "MEDIUM MATCH"

    return "LOW MATCH"


overall_status = score_status(
    overall_score
)


# ================================================================
# 26. STRUCTURED JSON
# ================================================================

evaluation_json = {

    "candidate": {
        "name": candidate_name,
        "phone": candidate_phone,
        "email": candidate_email
    },

    "job": {
        "job_title":
            jd_data["job_title"],

        "required_experience":
            jd_data["experience"],

        "required_skills":
            normalized_jd_skills
    },

    "candidate_fit": {
        "overall_match_score":
            overall_score,

        "overall_status":
            overall_status
    },

    "resume": {
        "sections":
            sections
    },

    "skills": {
        "candidate_skills":
            sorted(
                candidate_skills,
                key=lambda x: x.lower()
            ),

        "matched_required_skills":
            matched_required_skills,

        "missing_required_skills":
            missing_required_skills,

        "required_skill_score":
            skill_score,

        "skill_evidence":
            skill_evidence
    },

    "experience": {

        "required_experience":
            f"{required_experience_years} year",

        "experience_duration":
            experience_duration,

        "experience_eligibility":
            experience_eligibility,

        "experience_evidence_score":
            experience_score,

        "roles":
            experience_blocks,

        "technical_experience_text":
            technical_experience_text,

        "evidence":
            experience_results
    },

    "projects": {

        "total_projects":
            len(project_results),

        "projects":
            project_results,

        "project_match_score":
            project_score
    },

    "score_breakdown": {

        "skills": {
            "score": skill_score,
            "weight": SKILL_WEIGHT,
            "contribution":
                round(
                    skill_score * SKILL_WEIGHT,
                    2
                )
        },

        "experience": {
            "score": experience_score,
            "weight": EXPERIENCE_WEIGHT,
            "contribution":
                round(
                    experience_score * EXPERIENCE_WEIGHT,
                    2
                )
        },

        "projects": {
            "score": project_score,
            "weight": PROJECT_WEIGHT,
            "contribution":
                round(
                    project_score * PROJECT_WEIGHT,
                    2
                )
        }
    }
}


# ================================================================
# 27. HUMAN-READABLE REPORT
# ================================================================

print("\n\n")
print("=" * 80)
print("FINAL CANDIDATE EVALUATION")
print("=" * 80)


# ------------------------------------------------
# Candidate
# ------------------------------------------------

print("\nCANDIDATE INFORMATION")
print("-" * 80)

print("Name  :", candidate_name)
print("Phone :", candidate_phone)
print("Email :", candidate_email)


# ------------------------------------------------
# Job
# ------------------------------------------------

print("\nJOB INFORMATION")
print("-" * 80)

print(
    "Job Title:",
    jd_data["job_title"]
)

print(
    "Required Experience:",
    f"{required_experience_years} year"
)


# ------------------------------------------------
# Skills
# ------------------------------------------------

print("\nSKILL ANALYSIS")
print("-" * 80)

print("Candidate Skills:")

for skill in sorted(
    candidate_skills,
    key=lambda x: x.lower()
):

    print(
        f"  • {skill}"
    )

print("\nMatched Required Skills:")

for skill in matched_required_skills:

    print(
        f"  ✓ {skill}"
    )

    for item in skill_evidence[skill]:

        print(
            f"      Source: {item['source']}"
        )

        print(
            f"      Evidence: "
            f"{item['matched_text']}"
        )

print("\nMissing Required Skills:")

for skill in missing_required_skills:

    print(
        f"  ✗ {skill}"
    )

print(
    f"\nSkill Match Score: "
    f"{skill_score:.2f}%"
)


# ------------------------------------------------
# Experience
# ------------------------------------------------

print("\nEXPERIENCE ANALYSIS")
print("-" * 80)

print(
    "\nDetected Experience Roles:"
)

if experience_blocks:

    for index, block in enumerate(
        experience_blocks,
        start=1
    ):

        print(
            f"\n  Role {index}: "
            f"{block['role_title']}"
        )

        print(
            f"    Company: "
            f"{block['company']}"
        )

        print(
            f"    Dates: "
            f"{block['start']} - "
            f"{block['end']}"
        )

        print(
            f"    Classification: "
            f"{block['role_type']}"
        )

else:

    print("  No dated experience roles detected.")


print(
    "\nTechnical Experience Duration:"
)

if experience_duration["total_years"] is None:

    print("  Unknown")

else:

    print(
        f"  {experience_duration['total_years']} years "
        f"({experience_duration['total_months']} months)"
    )


print(
    "\nExperience Eligibility:",
    experience_eligibility
)


print(
    "\nTechnical Roles Considered:"
)

if experience_duration["technical_roles"]:

    for role in experience_duration[
        "technical_roles"
    ]:

        print(
            f"  ✓ {role['role']} | "
            f"{role['company']} | "
            f"{role['start']} - "
            f"{role['end']} | "
            f"{role['months']} months"
        )

else:

    print("  None")


print(
    "\nNon-Technical Roles Excluded:"
)

if experience_duration[
    "excluded_non_technical_roles"
]:

    for role in experience_duration[
        "excluded_non_technical_roles"
    ]:

        print(
            f"  - {role['role']} | "
            f"{role['company']} | "
            f"{role['start']} - "
            f"{role['end']}"
        )

else:

    print("  None")


print(
    "\nExperience Evidence:"
)

for requirement, result in (
    experience_results.items()
):

    print(
        f"  {requirement}: "
        f"{result['strength']}"
    )

    if result["evidence"]:

        print(
            "      Evidence:",
            ", ".join(
                result["evidence"]
            )
        )

print(
    f"\nExperience Evidence Score: "
    f"{experience_score:.2f}%"
)


# ------------------------------------------------
# Projects
# ------------------------------------------------

print("\nPROJECT ANALYSIS")
print("-" * 80)

print(
    "Total Projects Detected:",
    len(project_results)
)

# ------------------------------------------------
# Raw project-section verification.
# This makes it easy to diagnose extraction problems
# without changing the project score.
# ------------------------------------------------
print("\nRAW PROJECT SECTION LINES:")
print("-" * 80)

for raw_index, raw_project_line in enumerate(
    sections.get("Projects", []),
    start=1
):
    print(
        f"  [{raw_index}] {raw_project_line}"
    )

if not project_results:

    print(
        "\nWARNING: No projects detected."
    )

    print(
        "Raw Projects Section:"
    )

    for line in sections.get(
        "Projects",
        []
    ):

        print(
            "  ",
            line
        )

else:

    for index, project in enumerate(
        project_results,
        start=1
    ):

        print(
            f"\nProject {index}: "
            f"{project['title']}"
        )

        print(
            "Description:",
            project["description"]
        )

        print(
            "Project Match Score:",
            f"{project['project_match_score']:.2f}%"
        )

        for requirement, evidence in (
            project["evidence"].items()
        ):

            if evidence["strength"] != "NONE":

                print(
                    f"   {requirement}: "
                    f"{evidence['strength']}"
                )

                print(
                    "      Evidence:",
                    ", ".join(
                        evidence["evidence"]
                    )
                )

print(
    f"\nOverall Project Score: "
    f"{project_score:.2f}%"
)


# ------------------------------------------------
# Score Breakdown
# ------------------------------------------------

print("\nSCORE BREAKDOWN")
print("-" * 80)

print(
    f"Skills: "
    f"{skill_score:.2f}% × "
    f"{SKILL_WEIGHT} = "
    f"{skill_score * SKILL_WEIGHT:.2f}"
)

print(
    f"Experience: "
    f"{experience_score:.2f}% × "
    f"{EXPERIENCE_WEIGHT} = "
    f"{experience_score * EXPERIENCE_WEIGHT:.2f}"
)

print(
    f"Projects: "
    f"{project_score:.2f}% × "
    f"{PROJECT_WEIGHT} = "
    f"{project_score * PROJECT_WEIGHT:.2f}"
)


# ------------------------------------------------
# Final Result
# ------------------------------------------------

print("\nFINAL RESULT")
print("-" * 80)

print(
    f"Overall Match Score: "
    f"{overall_score:.2f}%"
)

print(
    "Overall Status:",
    overall_status
)


# ================================================================
# 28. SAVE JSON
# ================================================================

with open(
    RESULT_JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        evaluation_json,
        file,
        indent=4,
        ensure_ascii=False,
        default=str
    )


print("\nJSON result saved to:")
print(RESULT_JSON_PATH)

# ================================================================
# 29. STRUCTURED JSON OUTPUT
# ================================================================
#
# This is intentionally printed after the human-readable report
# so you can inspect the exact machine-readable result.
# ================================================================

print("\n" + "=" * 80)
print("STRUCTURED JSON OUTPUT")
print("=" * 80)

print(
    json.dumps(
        evaluation_json,
        indent=4,
        ensure_ascii=False,
        default=str
    )
)

print("\n" + "=" * 80)
print("EVALUATION COMPLETED")
print("=" * 80)
```

```python
# Interview Expert - Robust Resume/JD Matching Baseline
# Replace RESUME_PATH and JD_PATH below, then run this file.
#
# This version keeps the existing output format while fixing:
# 1. Skills polluted by project/experience text
# 2. Project detection for "Title | Tech | Year" formats
# 3. Internship/experience detection with year-only dates
# 4. Multiple experience entries
# 5. Python != automatic backend evidence
# 6. API usage != automatic REST API development
# 7. Java != JavaScript and SQL != MySQL
#
# Install:
# pip install pymupdf pytesseract pillow

from pathlib import Path
import json, re, sys
from datetime import date

try:
    import fitz
except ImportError:
    fitz = None

try:
    import pytesseract
    from PIL import Image
except ImportError:
    pytesseract = None
    Image = None

RESUME_PATH = Path(r"..\data\Sumit_JoshSoftware_resume.pdf")
JD_PATH = Path(r"..\data\sample_jd.txt")
OUTPUT_PATH = Path(r"..\data\candidate_evaluation_result.json")
TESSERACT_PATH = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

WEIGHTS = {"skills": 0.50, "experience": 0.30, "projects": 0.20}
HIGH_MATCH_THRESHOLD = 70.0
MEDIUM_MATCH_THRESHOLD = 40.0

DEFAULT_JD = """Job Title: Junior Full Stack Developer
Required Experience: 1 year
Required Skills: Java, Spring Boot, MySQL, HTML, CSS, JavaScript, React, Node.js
"""

SECTION_ALIASES = {
    "Professional Summary": {"professional summary","profile","career summary","summary","objective","career objective"},
    "Skills": {"skills","technical skills","core skills","key skills","technical expertise","technologies","tech stack","skills & technologies","skills and technologies","competencies"},
    "Projects": {"projects","project","academic projects","personal projects","key projects","selected projects","project experience"},
    "Internship & Experience": {"experience","work experience","professional experience","employment","employment history","internship","internships","internship experience","internship & experience","work history"},
    "Education": {"education","academic background","educational background","academic qualifications","qualifications"},
    "Certifications": {"certifications","certificates","licenses & certifications","licenses and certifications"},
    "Achievements": {"achievements","awards","honors","accomplishments"},
    "Leadership & Volunteering": {"leadership","leadership & volunteering","volunteering","volunteer experience","positions of responsibility","extracurricular","extra curricular"},
}

SKILL_ALIASES = {
    "React":["react","react.js","reactjs"], "Node.js":["node.js","nodejs","node js"],
    "Express.js":["express.js","expressjs","express js"], "JavaScript":["javascript"],
    "TypeScript":["typescript"], "Spring Boot":["spring boot","springboot"],
    "MySQL":["mysql"], "HTML":["html","html5"], "CSS":["css","css3"],
    "Python":["python"], "FastAPI":["fastapi"], "MongoDB":["mongodb","mongo db"],
    "PostgreSQL":["postgresql","postgres"], "SQL":["sql","structured query language"],
    "REST API":["rest api","restful api"], "AWS":["aws","amazon web services"],
    "Git":["git"], "GitHub":["github"], "Next.js":["next.js","nextjs"],
    "Tailwind CSS":["tailwind css","tailwindcss"], "C++":["c++"], "PHP":["php"],
    "Java":["java"], "SQLite":["sqlite"], "Firebase":["firebase"], "Convex":["convex"],
    "Power BI":["power bi"], "Tableau":["tableau"], "Power Query":["power query"],
    "DAX":["dax"], "Excel":["excel","microsoft excel"], "Pandas":["pandas"],
    "NumPy":["numpy"], "Matplotlib":["matplotlib"], "Seaborn":["seaborn"],
    "Scikit-learn":["scikit-learn","sklearn"], "Jupyter Notebook":["jupyter notebook","jupiter notebook"],
    "Google Colab":["google colab"], "VS Code":["vs code","visual studio code"],
    "Figma":["figma"], "Postman":["postman"], "Docker":["docker"], "Regex":["regex"],
}

PROJECT_TECHS = list(SKILL_ALIASES.keys()) + ["Razorpay","Twilio","Google Maps API","LLM","LLMs","Telegram"]

TECH_ROLE_TERMS = ["developer","engineer","intern","software","web","full stack","frontend","backend",
                   "data scientist","data analyst","machine learning","ai developer","cloud","devops",
                   "qa tester","qa engineer","programmer","cybersecurity","technical intern"]
NONTECH_ROLE_TERMS = ["event co-head","vice president","volunteer","student coordinator","cultural secretary",
                      "club coordinator","marketing coordinator","event coordinator"]

def clean_line(s):
    s = s.replace("\xa0"," ").replace("\u200b","").replace("\uf0b7","•")
    return re.sub(r"\s+"," ",s).strip()

def clean_text(s):
    return "\n".join(clean_line(x) for x in s.splitlines() if clean_line(x))

def strip_bullet(s):
    return re.sub(r"^[\s•●▪▫◦‣⁃∙·*\-–—]+","",s).strip()

def norm(s):
    return re.sub(r"\s+"," ",re.sub(r"[^a-z0-9+#.\- ]+"," ",s.lower())).strip()

def unique(items):
    out=[]; seen=set()
    for x in items:
        k=norm(x)
        if k and k not in seen: seen.add(k); out.append(x)
    return out

def configure_ocr():
    if pytesseract and Path(TESSERACT_PATH).exists():
        pytesseract.pytesseract.tesseract_cmd=TESSERACT_PATH

def extract_pdf(path):
    if fitz is None: raise RuntimeError("Install PyMuPDF: pip install pymupdf")
    if not path.exists(): raise FileNotFoundError(path)
    configure_ocr()
    doc=fitz.open(str(path)); pages=[p.get_text("text") or "" for p in doc]
    chars=sum(map(len,pages))
    if chars >= 300:
        return clean_text("\n".join(pages)), len(doc), chars, False
    if not pytesseract or not Image:
        raise RuntimeError("PDF text is insufficient; install pytesseract and pillow.")
    out=[]
    for p in doc:
        pix=p.get_pixmap(matrix=fitz.Matrix(2,2),alpha=False)
        img=Image.frombytes("RGB",[pix.width,pix.height],pix.samples)
        out.append(pytesseract.image_to_string(img))
    return clean_text("\n".join(out)), len(doc), chars, True

def heading_key(line):
    x=strip_bullet(line).strip(" :|-–—").lower()
    for section, aliases in SECTION_ALIASES.items():
        if x in aliases: return section
    return None

def detect_sections(text):
    sections={}; current=None
    for line in text.splitlines():
        line=clean_line(line)
        if not line: continue
        h=heading_key(line)
        if h:
            current=h; sections.setdefault(h,[]); continue
        if current: sections[current].append(line)
    return sections

def skill_pattern(alias):
    a=norm(alias); e=re.escape(a).replace(r"\ ",r"\s*")
    if a=="java": return r"(?<![A-Za-z0-9_])java(?!script)(?![A-Za-z0-9_])"
    if a=="sql": return r"(?<![A-Za-z0-9_])sql(?![A-Za-z0-9_])"
    return rf"(?<![A-Za-z0-9_]){e}(?![A-Za-z0-9_])"

def skill_match(skill,text):
    return any(re.search(skill_pattern(a),norm(text),re.I) for a in SKILL_ALIASES.get(skill,[skill]))

def canonical_skill(x):
    for c, aliases in SKILL_ALIASES.items():
        if norm(x) in {norm(a) for a in aliases}: return c
    return None

def extract_skills(lines):
    found=[]
    # Only scan the actual Skills section.
    vocabulary=list(SKILL_ALIASES.keys()) + ["Razorpay","Twilio","Google Maps API"]
    for raw in lines:
        line=strip_bullet(raw)
        line=re.sub(r"^(languages?|programming(?: languages?)?|frontend(?: development)?|backend(?: development)?|"
                    r"databases?|tools?|libraries?|frameworks?|cloud(?: & devops)?|technologies|"
                    r"visualization tools?|core capabilities?|testing(?: & tools?)?|communication skills?)\s*:\s*",
                    "",line,flags=re.I)
        for s in vocabulary:
            if skill_match(s,line): found.append(s)
        # exact comma/pipe items only; reject sentences
        for piece in re.split(r"[,;|]",line):
            p=re.sub(r"\s*\([^)]*\)","",strip_bullet(piece)).strip()
            if not p or len(p)>35: continue
            if re.search(r"\b(built|developed|implemented|performed|designed|configured|project|dashboard|model)\b",p,re.I): continue
            c=canonical_skill(p)
            if c: found.append(c)
    return unique(found)

def parse_jd(text):
    text=clean_text(text)
    tm=re.search(r"(?:job\s*title|position|role)\s*:\s*(.+)",text,re.I)
    em=re.search(r"(\d+(?:\.\d+)?)\s*\+?\s*years?",text,re.I)
    sm=re.search(r"(?:required\s+skills?|must\s+have|technical\s+skills?)\s*:\s*(.+)",text,re.I)
    catalog=list(SKILL_ALIASES.keys())
    source=sm.group(1) if sm else text
    skills=[s for s in catalog if skill_match(s,source)]
    if sm:
        explicit=[strip_bullet(x).strip() for x in re.split(r"[,;|\n]",sm.group(1)) if 1<=len(strip_bullet(x).strip())<=60]
        if explicit: skills=unique(explicit)
    return {"job_title":tm.group(1).strip() if tm else "Unknown",
            "required_experience":{"min_years":float(em.group(1)) if em else None,"max_years":None},
            "required_skills":skills or ["Java","Spring Boot","MySQL","HTML","CSS","JavaScript","React","Node.js"]}

MONTHS=r"Jan(?:uary)?|Feb(?:ruary)?|Mar(?:ch)?|Apr(?:il)?|May|Jun(?:e)?|Jul(?:y)?|Aug(?:ust)?|Sep(?:t(?:ember)?)?|Oct(?:ober)?|Nov(?:ember)?|Dec(?:ember)?"
DATE_RANGE=re.compile(rf"((?:{MONTHS})\s+\d{{4}}|\d{{1,2}}[/-]\d{{4}}|\d{{4}})\s*(?:-|–|—|to)\s*((?:{MONTHS})\s+\d{{4}}|\d{{1,2}}[/-]\d{{4}}|\d{{4}}|Present|Current|Now)",re.I)
YEAR_ONLY=re.compile(r"^(19|20)\d{2}$")

def parse_part(v):
    v=v.strip()
    if re.fullmatch(r"present|current|now",v,re.I): return date.today().replace(day=1),True
    m=re.match(rf"({MONTHS})\s+(\d{{4}})",v,re.I)
    if m:
        mm={"jan":1,"feb":2,"mar":3,"apr":4,"may":5,"jun":6,"jul":7,"aug":8,"sep":9,"oct":10,"nov":11,"dec":12}[m.group(1).lower()[:3]]
        return date(int(m.group(2)),mm,1),False
    m=re.fullmatch(r"(\d{1,2})[/-](\d{4})",v)
    if m:return date(int(m.group(2)),int(m.group(1)),1),False
    if re.fullmatch(r"(19|20)\d{2}",v): return date(int(v),1,1),False
    return None,False

def months_between(a,b): return max(0,(b.year-a.year)*12+b.month-a.month+1)

def classify_role(role,body):
    r=norm(role); b=norm(body)
    if any(x in r for x in NONTECH_ROLE_TERMS): return "NON_TECHNICAL"
    if any(x in r for x in TECH_ROLE_TERMS): return "TECHNICAL"
    signals=["developed","deployed","website","web application","api","database","python","java","javascript",
             "react","node.js","html","css","fastapi","aws","testing","machine learning","cloud"]
    return "TECHNICAL" if sum(x in b for x in signals)>=2 else "OTHER"

def likely_title(line):
    x=strip_bullet(line)
    if not x or len(x)>100 or heading_key(x): return False
    if re.match(r"^(developed|built|implemented|configured|performed|designed|assisted|participated|worked|leading|coordinated|contributed|created|maintained)\b",x,re.I): return False
    if re.search(r"(developer|engineer|intern|analyst|scientist|manager|specialist|architect|tester|programmer|consultant|trainee|lead|associate|executive|coordinator|president|co-head)",x,re.I): return True
    return len(x.split())<=12

def detect_experience(lines):
    roles=[]; i=0
    while i<len(lines):
        line=strip_bullet(lines[i]); m=DATE_RANGE.search(line)
        year=YEAR_ONLY.match(line)
        if not m and not year: i+=1; continue
        candidates=[strip_bullet(lines[k]) for k in range(max(0,i-5),i) if strip_bullet(lines[k])]
        role=next((x for x in candidates if any(t in norm(x) for t in TECH_ROLE_TERMS+NONTECH_ROLE_TERMS)),None)
        if not role: role=next((x for x in candidates if likely_title(x)),None)
        company=next((x for x in candidates if x!=role),None)
        body=[]; j=i+1
        while j<len(lines):
            nxt=strip_bullet(lines[j])
            if not nxt: j+=1; continue
            if DATE_RANGE.search(nxt) or YEAR_ONLY.match(nxt): break
            if j>i+1 and likely_title(nxt) and any(t in norm(nxt) for t in TECH_ROLE_TERMS+NONTECH_ROLE_TERMS): break
            body.append(nxt); j+=1
        if m:
            sd,cur=parse_part(m.group(1)); ed,ecur=parse_part(m.group(2))
            date_text=m.group(0); end=ed.isoformat() if ed else None; start=sd.isoformat() if sd else None; certain=True; current=ecur
        else:
            y=int(year.group(0)); date_text=year.group(0); start=f"{y}-01-01"; end=None; certain=False; current=False
        body_text=" ".join(body)
        roles.append({"role_title":role or "Unknown Role","company":company or "Unknown Company","date_text":date_text,
                      "start_date":start,"end_date":end,"is_current":current,"duration_certain":certain,
                      "role_type":classify_role(role or "",body_text),"body":body_text})
        i=max(i+1,j)
    out=[]; seen=set()
    for r in roles:
        key=(norm(r["role_title"]),norm(r["company"]),r["date_text"])
        if key not in seen: seen.add(key); out.append(r)
    return out

def experience_evidence(text):
    t=norm(text)
    rules={
      "full stack development":[r"\bfull[- ]stack\b",r"\bmern\b",r"\bmean\b",r"\bend[- ]to[- ]end\b"],
      "web application development":[r"\bweb application\b",r"\bweb app\b",r"\bwebsite\b",r"\bweb platform\b",r"\bweb portal\b"],
      "frontend development":[r"\bfrontend\b",r"\bfront end\b",r"\breact\b",r"\bhtml\b",r"\bcss\b"],
      "backend development":[r"\bbackend\b",r"\bback end\b",r"\bserver[- ]side\b",r"\bfastapi\b",r"\bflask\b",r"\bdjango\b",r"\bexpress\.js\b"],
      "REST API development":[r"\bdeveloped\b.{0,60}\brest(?:ful)?\s+api\b",r"\bimplemented\b.{0,60}\brest(?:ful)?\s+api\b",r"\bbuilt\b.{0,60}\brest(?:ful)?\s+api\b",r"\brest api development\b"],
      "database development":[r"\bdatabase\b",r"\bsqlite\b",r"\bmysql\b",r"\bmongodb\b",r"\bpostgresql\b",r"\bsql\b",r"\bfirebase\b"]
    }
    result={}
    for cat,pats in rules.items():
        hit=next((p for p in pats if re.search(p,t,re.I)),None)
        result[cat]={"evidence":[text[:220]] if hit else [],"strength":"STRONG" if hit else "NONE","numeric_score":1.0 if hit else 0.0}
    return result

def score_projects(lines):
    projects=[]; current=None
    for raw in lines:
        line=clean_line(raw)
        if not line: continue
        x=strip_bullet(line)
        strong=False
        if "|" in x:
            parts=x.split("|")
            strong=len(parts)>=2 and (any(skill_match(s," ".join(parts[1:])) for s in PROJECT_TECHS) or re.search(r"\b(19|20)\d{2}\b",x))
        if re.search(r"\bproject\b",x,re.I): strong=True
        if re.search(r"\s[-–—:]\s",x):
            left,right=re.split(r"\s[-–—:]\s",x,maxsplit=1)
            strong=strong or (len(left.split())<=16 and any(skill_match(s,right) for s in PROJECT_TECHS))
        if re.search(r"\b(19|20)\d{2}\b",x) and len(re.sub(r"\b(19|20)\d{2}\b","",x).split())<=16: strong=True
        if not re.match(r"^(built|developed|implemented|designed|performed|configured|created|deployed|used)\b",x,re.I) and strong:
            if current: projects.append(current)
            title=x; year=(re.search(r"\b(19|20)\d{2}\b",x).group(0) if re.search(r"\b(19|20)\d{2}\b",x) else None)
            noyear=re.sub(r"\b(19|20)\d{2}\b","",x).strip()
            parts=[p.strip() for p in noyear.split("|")]
            title=parts[0].strip(" -–—:") if len(parts)>1 else noyear
            techs=[]
            if len(parts)>1: techs=[s for s in PROJECT_TECHS if skill_match(s," ".join(parts[1:]))]
            current={"title":title,"technologies":unique(techs),"year":year,"description_lines":[]}
        elif current:
            current["description_lines"].append(x)
    if current: projects.append(current)
    final=[]
    for p in projects:
        p["description"]=" ".join(p.pop("description_lines")).strip()
        p["technologies"]=unique(p["technologies"]+[s for s in PROJECT_TECHS if skill_match(s,p["description"])])
        t=norm(p["title"])
        if t and t not in {"project","tech","technologies"}: final.append(p)
    # score using explicit evidence, not language/library names alone as backend
    for p in final:
        t=norm(p["title"]+" "+p["description"]+" "+" ".join(p["technologies"]))
        e={}
        e["full stack development"]={"evidence":["explicit full-stack/end-to-end"] if re.search(r"\bfull[- ]stack\b|\bmern\b|\bmean\b",t) else [],"strength":"STRONG" if re.search(r"\bfull[- ]stack\b|\bmern\b|\bmean\b",t) else "NONE","numeric_score":1.0 if re.search(r"\bfull[- ]stack\b|\bmern\b|\bmean\b",t) else 0.0}
        e["web application development"]={"evidence":["web application/platform/website"] if re.search(r"\bweb application\b|\bweb app\b|\bweb platform\b|\bwebsite\b",t) else [],"strength":"STRONG" if re.search(r"\bweb application\b|\bweb app\b|\bweb platform\b|\bwebsite\b",t) else "NONE","numeric_score":1.0 if re.search(r"\bweb application\b|\bweb app\b|\bweb platform\b|\bwebsite\b",t) else 0.0}
        e["frontend development"]={"evidence":["frontend/UI technology"] if re.search(r"\bfrontend\b|\breact\b|\breact native\b|\bhtml\b|\bcss\b",t) else [],"strength":"STRONG" if re.search(r"\bfrontend\b|\breact\b|\breact native\b|\bhtml\b|\bcss\b",t) else "NONE","numeric_score":1.0 if re.search(r"\bfrontend\b|\breact\b|\breact native\b|\bhtml\b|\bcss\b",t) else 0.0}
        e["backend development"]={"evidence":["explicit backend/server technology"] if re.search(r"\bbackend\b|\bback end\b|\bserver[- ]side\b|\bfastapi\b|\bdjango\b|\bflask\b|\bexpress\.js\b",t) else [],"strength":"STRONG" if re.search(r"\bbackend\b|\bback end\b|\bserver[- ]side\b|\bfastapi\b|\bdjango\b|\bflask\b|\bexpress\.js\b",t) else "NONE","numeric_score":1.0 if re.search(r"\bbackend\b|\bback end\b|\bserver[- ]side\b|\bfastapi\b|\bdjango\b|\bflask\b|\bexpress\.js\b",t) else 0.0}
        rest=re.search(r"\b(developed|implemented|built)\b.{0,60}\brest(?:ful)?\s+api\b",t)
        api=bool(re.search(r"\bapi\b",t))
        e["REST API development"]={"evidence":["explicit REST API development"] if rest else (["API mentioned; REST development not explicit"] if api else []),"strength":"STRONG" if rest else ("PARTIAL" if api else "NONE"),"numeric_score":1.0 if rest else (0.5 if api else 0.0)}
        db=bool(re.search(r"\bdatabase\b|\bsqlite\b|\bmysql\b|\bmongodb\b|\bpostgresql\b|\bsql\b|\bfirebase\b",t))
        e["database development"]={"evidence":["database technology"] if db else [],"strength":"STRONG" if db else "NONE","numeric_score":1.0 if db else 0.0}
        p["evidence"]=e
        p["project_match_score"]=round(sum(v["numeric_score"] for v in e.values())/6*100,2)
    return final

def main():
    print("="*80); print("RESUME–JD MATCHING SYSTEM"); print("="*80)
    print("\nResume path:",RESUME_PATH,"\nFile exists:",RESUME_PATH.exists())
    text,pages,chars,ocr=extract_pdf(RESUME_PATH)
    print("\nSelectable PDF text characters:",chars,"\nPDF pages:",pages)
    print("\nOCR used." if ocr else "\nSelectable text detected.\nOCR not required.")
    print("\nFinal extracted characters:",len(text))
    sections=detect_sections(text)
    print("\n"+"="*80); print("DETECTED RESUME SECTIONS"); print("="*80)
    for s,l in sections.items(): print(f"{s}: {len(l)} lines")
    jd=parse_jd(JD_PATH.read_text(encoding="utf-8",errors="ignore") if JD_PATH.exists() else DEFAULT_JD)
    candidate_skills=extract_skills(sections.get("Skills",[]))
    matched=[s for s in jd["required_skills"] if any(skill_match(s,c) or norm(s)==norm(c) for c in candidate_skills)]
    missing=[s for s in jd["required_skills"] if s not in matched]
    skill_score=round(len(matched)/len(jd["required_skills"])*100,2) if jd["required_skills"] else 0
    skill_evidence={s:[{"source":src,"matched_text":s,"context":ln} for src,ls in sections.items() for ln in ls if skill_match(s,ln)] for s in jd["required_skills"]}
    roles=detect_experience(sections.get("Internship & Experience",[]))
    tech=[r for r in roles if r["role_type"]=="TECHNICAL"]; excluded=[r for r in roles if r["role_type"]!="TECHNICAL"]
    certain_months=0
    for r in tech:
        if r["duration_certain"] and r["end_date"]:
            certain_months+=months_between(date.fromisoformat(r["start_date"]),date.fromisoformat(r["end_date"]))
    exp_years=certain_months/12 if certain_months else None
    req=jd["required_experience"]["min_years"]
    eligibility=("MEETS_REQUIREMENT" if exp_years is not None and req is not None and exp_years>=req else
                 "PARTIAL" if tech else ("DOES_NOT_MEET" if roles else "UNKNOWN"))
    exp_text=" ".join(r["body"] for r in tech)
    exp_ev=experience_evidence(exp_text)
    exp_score=round(sum(v["numeric_score"] for v in exp_ev.values())/6*100,2)
    projects=score_projects(sections.get("Projects",[]))
    project_score=round(sum(p["project_match_score"] for p in projects)/len(projects),2) if projects else 0
    final=round(skill_score*.5+exp_score*.3+project_score*.2,2)
    status="HIGH MATCH" if final>=70 else ("MEDIUM MATCH" if final>=40 else "LOW MATCH")
    print("\n"+"="*80); print("FINAL CANDIDATE EVALUATION"); print("="*80)
    name=(re.search(r"(?m)^([A-Z][A-Za-z .'-]{2,60})$",text).group(1).strip() if re.search(r"(?m)^([A-Z][A-Za-z .'-]{2,60})$",text) else "Unknown")
    email=(re.search(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",text).group(0) if re.search(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}",text) else "Unknown")
    phone=(re.search(r"(?:\+91[\s-]?)?\d{10}",text).group(0) if re.search(r"(?:\+91[\s-]?)?\d{10}",text) else "Unknown")
    print("\nCANDIDATE INFORMATION\n"+"-"*80); print("Name  :",name); print("Phone :",phone); print("Email :",email)
    print("\nJOB INFORMATION\n"+"-"*80); print("Job Title:",jd["job_title"]); print("Required Experience:",f"{req:g} year" if req else "Not specified")
    print("\nSKILL ANALYSIS\n"+"-"*80); print("Candidate Skills:"); [print("  •",s) for s in candidate_skills]
    print("\nMatched Required Skills:"); [print("  ✓",s) for s in matched]
    print("\nMissing Required Skills:"); [print("  ✗",s) for s in missing]
    print(f"\nSkill Match Score: {skill_score:.2f}%")
    print("\nEXPERIENCE ANALYSIS\n"+"-"*80); print("\nDetected Experience Roles:")
    for i,r in enumerate(roles,1): print(f"\n  Role {i}: {r['role_title']}\n    Company: {r['company']}\n    Dates: {r['date_text']}\n    Classification: {r['role_type']}")
    if not roles: print("  No experience roles detected.")
    print("\nTechnical Experience Duration:",f"{exp_years:.2f} years" if exp_years is not None else "Not confidently determinable from resume dates.")
    print("\nExperience Eligibility:",eligibility)
    print("\nTechnical Roles Considered:"); [print("  -",r["role_title"],"|",r["company"],"|",r["date_text"]) for r in tech] or print("  None")
    print("\nNon-Technical Roles Excluded:"); [print("  -",r["role_title"],"|",r["company"],"|",r["date_text"]) for r in excluded] or print("  None")
    print("\nExperience Evidence:"); [print(f"  {k}: {v['strength']}") for k,v in exp_ev.items()]
    print(f"\nExperience Evidence Score: {exp_score:.2f}%")
    print("\nPROJECT ANALYSIS\n"+"-"*80); print("Total Projects Detected:",len(projects))
    print("\nRAW PROJECT SECTION LINES:\n"+"-"*80); [print(f"  [{i}] {x}") for i,x in enumerate(sections.get("Projects",[]),1)]
    for i,p in enumerate(projects,1):
        print(f"\nProject {i}: {p['title']}\nDescription: {p['description']}\nProject Match Score: {p['project_match_score']:.2f}%")
        [print(f"   {k}: {v['strength']}",*(f"\n      Evidence: {e}" for e in v["evidence"])) for k,v in p["evidence"].items()]
    print(f"\nOverall Project Score: {project_score:.2f}%")
    print("\nSCORE BREAKDOWN\n"+"-"*80); print(f"Skills: {skill_score:.2f}% × 0.5 = {skill_score*.5:.2f}"); print(f"Experience: {exp_score:.2f}% × 0.3 = {exp_score*.3:.2f}"); print(f"Projects: {project_score:.2f}% × 0.2 = {project_score*.2:.2f}")
    print("\nFINAL RESULT\n"+"-"*80); print(f"Overall Match Score: {final:.2f}%"); print("Overall Status:",status)
    result={"candidate":{"name":name,"phone":phone,"email":email},"job":jd,
            "candidate_fit":{"overall_match_score":final,"overall_status":status},
            "resume":{"sections":sections},
            "skills":{"candidate_skills":candidate_skills,"matched_required_skills":matched,"missing_required_skills":missing,"required_skill_score":skill_score,"skill_evidence":skill_evidence},
            "experience":{"required_experience":f"{req:g} year" if req else None,"experience_duration":{"status":eligibility,"total_months":certain_months or None,"total_years":round(exp_years,2) if exp_years else None,"technical_roles":tech,"excluded_non_technical_roles":excluded},"experience_eligibility":eligibility,"experience_evidence_score":exp_score,"roles":roles,"technical_experience_text":exp_text,"evidence":exp_ev},
            "projects":{"total_projects":len(projects),"projects":projects,"project_match_score":project_score},
            "score_breakdown":{"skills":{"score":skill_score,"weight":.5,"contribution":round(skill_score*.5,2)},"experience":{"score":exp_score,"weight":.3,"contribution":round(exp_score*.3,2)},"projects":{"score":project_score,"weight":.2,"contribution":round(project_score*.2,2)}}}
    OUTPUT_PATH.parent.mkdir(parents=True,exist_ok=True); OUTPUT_PATH.write_text(json.dumps(result,indent=4,ensure_ascii=False,default=str),encoding="utf-8")
    print("\nJSON result saved to:",OUTPUT_PATH)
    print("\n"+"="*80); print("STRUCTURED JSON OUTPUT"); print("="*80); print(json.dumps(result,indent=4,ensure_ascii=False,default=str))
    print("\n"+"="*80); print("EVALUATION COMPLETED"); print("="*80)

if __name__=="__main__": main()
```

```python
"""
Interview Expert
AI-Powered Resume–JD Matching and Candidate Evaluation System

V9 — Accuracy-focused baseline.

Important:
- This is a rule-based / heuristic matching engine, NOT a trained ML model.
- It extracts resume/JD text, sections, skills, experience and projects.
- It keeps direct skills, project evidence, experience evidence and
  certification evidence separate.
- It does not invent experience, skills, dates, companies, or project details.
"""

from __future__ import annotations

import argparse
import json
import re
from dataclasses import dataclass, asdict
from pathlib import Path
from typing import Dict, List, Optional, Tuple

# Optional PDF/OCR dependencies.
try:
    import fitz  # PyMuPDF
except ImportError:
    fitz = None

try:
    import pytesseract
    from PIL import Image
except ImportError:
    pytesseract = None
    Image = None


# ---------------------------------------------------------------------------
# CONFIGURATION
# ---------------------------------------------------------------------------

TESSERACT_PATH = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

if pytesseract is not None:
    try:
        pytesseract.pytesseract.tesseract_cmd = TESSERACT_PATH
    except Exception:
        pass


SECTION_ALIASES = {
    "Professional Summary": [
        "professional summary", "summary", "profile", "objective"
    ],
    "Education": [
        "education", "academic background", "educational qualification"
    ],
    "Projects": [
        "projects", "academic projects", "personal projects", "project"
    ],
    "Internship & Experience": [
        "internship & experience", "internships", "internship",
        "experience", "work experience", "professional experience"
    ],
    "Skills": [
        "skills", "technical skills", "skills & technologies",
        "technical expertise"
    ],
    "Certifications": [
        "certifications", "certificates", "courses & certifications"
    ],
}

# Canonical aliases. Matching is intentionally conservative.
SKILL_ALIASES = {
    "Python": [r"\bpython\b"],
    "JavaScript": [r"\bjavascript\b", r"\bjava\s*script\b"],
    "C++": [r"(?<![A-Za-z])c\+\+(?![A-Za-z])"],
    # Do NOT use a naked \bC\b across the whole resume.
    # It creates false positives such as S.S.C.
    "C": [r"(?<![A-Za-z.])c(?![A-Za-z.])"],
    "Java": [r"(?<!script\s)(?<![A-Za-z])java(?!script|[A-Za-z])"],
    "HTML": [r"\bhtml(?:5)?\b"],
    "CSS": [r"\bcss(?:3)?\b"],
    "React": [r"\breact(?:\.js|js)?\b"],
    "React Native": [r"\breact\s+native\b"],
    "Next.js": [r"\bnext(?:\.js|js)\b"],
    "Node.js": [r"\bnode(?:\.js|js)\b"],
    "Express.js": [r"\bexpress(?:\.js|js)\b"],
    "SQL": [r"(?<![A-Za-z])sql(?![A-Za-z])"],
    "MySQL": [r"\bmysql\b"],
    "MongoDB": [r"\bmongodb\b"],
    "Firebase": [r"\bfirebase\b"],
    "Git": [r"(?<![A-Za-z])git(?!hub|lab|[A-Za-z])"],
    "GitHub": [r"\bgithub\b"],
    "Docker": [r"\bdocker\b"],
    "REST API": [
        r"\brest\s+apis?\b",
        r"\brestful\s+apis?\b",
        r"\brest\s*api\b",
    ],
    "Flask": [r"\bflask\b"],
    "Django": [r"\bdjango\b"],
    "FastAPI": [r"\bfastapi\b"],
    "AWS": [r"\baws\b"],
    "Figma": [r"\bfigma\b"],
    "Tailwind CSS": [r"\btailwind(?:\s+css)?\b"],
    "Razorpay": [r"\brazorpay\b"],
    "Twilio": [r"\btwilio\b"],
    "Google Maps API": [r"\bgoogle\s+maps\s+api\b"],
    "Convex": [r"\bconvex\b"],
    "LLMs": [r"\bllms?\b", r"\blarge\s+language\s+models?\b"],
    "AST": [r"\bast\b", r"\babstract\s+syntax\s+tree\b"],
}

SOFT_SKILL_ALIASES = {
    "Communication Skills": [
        r"\bcommunication\s+skills?\b",
        r"\bcommunicat(?:e|ed|ing|ion)\b",
        r"\bcoordinat(?:e|ed|ing)\b",
        r"\bcollaborat(?:e|ed|ing|ion)\b",
        r"\bwork(?:ed|ing)?\s+with\b",
        r"\binteract(?:ed|ing)\s+with\b",
        r"\bpresent(?:ed|ing)\b",
    ],
    "Leadership": [
        r"\blead(?:ing|ership)?\b",
        r"\bmanaged?\b",
        r"\bvice president\b",
        r"\bco-?head\b",
    ],
    "Teamwork": [
        r"\bteam(?:s|work)?\b",
        r"\bcollaborat(?:e|ed|ing|ion)\b",
        r"\bcoordinat(?:e|ed|ing)\b",
    ],
}

ACTION_VERBS = {
    "built", "developed", "implemented", "created", "designed",
    "deployed", "refactored", "analyzed", "analysed", "integrated",
    "engineered", "programmed", "constructed", "using", "used",
    "worked", "solved", "improved", "achieved", "participating",
    "assisting", "leading", "coordinating", "managed", "contributed",
}

ROLE_KEYWORDS = {
    "intern", "internship", "developer", "engineer", "software",
    "analyst", "programmer", "trainee", "research", "developer",
    "web", "backend", "frontend", "full stack", "ai", "data",
}

TECH_ROLE_KEYWORDS = {
    "intern", "internship", "developer", "engineer", "software",
    "programmer", "developer", "web", "backend", "frontend",
    "full stack", "ai", "data", "machine learning", "ml",
}

NON_TECH_ROLE_KEYWORDS = {
    "president", "vice president", "co-head", "head", "coordinator",
    "event", "cultural", "club", "committee", "volunteer",
}


# ---------------------------------------------------------------------------
# DATA CLASSES
# ---------------------------------------------------------------------------

@dataclass
class ExperienceRole:
    role_title: str
    company: str = "Unknown Company"
    date_text: str = "Not specified"
    start_date: Optional[str] = None
    end_date: Optional[str] = None
    is_current: bool = False
    duration_certain: bool = False
    role_type: str = "OTHER"
    body: str = ""


@dataclass
class Project:
    title: str
    technologies: List[str]
    year: Optional[str]
    description: str


# ---------------------------------------------------------------------------
# GENERAL TEXT HELPERS
# ---------------------------------------------------------------------------

def clean_text(text: str) -> str:
    text = text.replace("\u2013", "–").replace("\u2014", "–")
    text = text.replace("\u00a0", " ")
    text = re.sub(r"[ \t]+", " ", text)
    return text.strip()


def clean_line(line: str) -> str:
    return clean_text(line).strip()


def strip_bullet(line: str) -> str:
    return re.sub(r"^\s*(?:[•●▪◦■*-]|\d+[.)])\s*", "", line).strip()


def is_year_only(line: str) -> bool:
    return bool(re.fullmatch(r"\s*(?:19|20)\d{2}\s*", line))


def is_date_line(line: str) -> bool:
    s = line.lower().strip()
    if is_year_only(s):
        return True
    return bool(re.search(
        r"\b(?:jan|feb|mar|apr|may|jun|jul|aug|sep|sept|oct|nov|dec)"
        r"[a-z]*\.?\s+(?:19|20)\d{2}\b", s
    )) or bool(re.search(r"\b(?:19|20)\d{2}\s*[–-]\s*(?:(?:19|20)\d{2}|present|current)\b", s))


def parse_date_range(date_text: str) -> Tuple[Optional[str], Optional[str], bool]:
    s = clean_text(date_text)
    if not s or s == "Not specified":
        return None, None, False

    year = re.fullmatch(r"(19|20)\d{2}", s)
    if year:
        return f"{s}-01-01", None, False

    months = {
        "jan": 1, "january": 1, "feb": 2, "february": 2,
        "mar": 3, "march": 3, "apr": 4, "april": 4,
        "may": 5, "jun": 6, "june": 6, "jul": 7, "july": 7,
        "aug": 8, "august": 8, "sep": 9, "sept": 9, "september": 9,
        "oct": 10, "october": 10, "nov": 11, "november": 11,
        "dec": 12, "december": 12,
    }

    parts = re.split(r"\s*[–-]\s*", s)
    if len(parts) != 2:
        m = re.search(r"\b((?:19|20)\d{2})\b", s)
        return (f"{m.group(1)}-01-01", None, False) if m else (None, None, False)

    def parse_one(value: str) -> Optional[str]:
        value = value.strip().lower()
        m = re.search(r"(19|20)\d{2}", value)
        if not m:
            return None
        y = int(m.group(0))
        mm = 1
        for name, num in months.items():
            if re.search(rf"\b{name}\b", value):
                mm = num
                break
        return f"{y:04d}-{mm:02d}-01"

    start = parse_one(parts[0])
    if re.search(r"\b(?:present|current)\b", parts[1].lower()):
        return start, None, True
    end = parse_one(parts[1])
    return start, end, bool(start and end)


def contains_skill(text: str, skill: str) -> bool:
    for pattern in SKILL_ALIASES.get(skill, []):
        if re.search(pattern, text, flags=re.IGNORECASE):
            return True
    return False


def normalize_skill(skill: str) -> str:
    s = clean_text(skill).lower().rstrip(".")
    aliases = {
        "react.js": "react",
        "reactjs": "react",
        "nodejs": "node.js",
        "node.js": "node.js",
        "expressjs": "express.js",
        "express.js": "express.js",
        "rest apis": "rest api",
        "rest api": "rest api",
        "restful api": "rest api",
        "restful apis": "rest api",
        "mysql": "mysql",
        "sql": "sql",
        "c++": "c++",
        "communication skills": "communication skills",
    }
    return aliases.get(s, s)


def evidence_context(text: str, start: int, end: int, radius: int = 180) -> str:
    left = max(0, start - radius)
    right = min(len(text), end + radius)
    return clean_text(text[left:right])


def remove_academic_abbreviations(text: str) -> str:
    # Critical fix: S.S.C. and H.S.C. must never produce a standalone C skill.
    text = re.sub(r"\bS\.S\.C\.\b", " ", text, flags=re.IGNORECASE)
    text = re.sub(r"\bH\.S\.C\.\b", " ", text, flags=re.IGNORECASE)
    return text


# ---------------------------------------------------------------------------
# PDF / OCR
# ---------------------------------------------------------------------------

def extract_pdf_text(pdf_path: Path) -> Tuple[str, int, bool]:
    if fitz is None:
        raise RuntimeError("PyMuPDF is not installed. Run: pip install pymupdf")

    doc = fitz.open(pdf_path)
    pages = len(doc)
    selectable = "\n".join(page.get_text("text") for page in doc).strip()

    if selectable:
        return selectable, pages, False

    if pytesseract is None or Image is None:
        return selectable, pages, True

    ocr_pages = []
    for page in doc:
        pix = page.get_pixmap(matrix=fitz.Matrix(2, 2), alpha=False)
        img = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
        ocr_pages.append(pytesseract.image_to_string(img))

    return "\n".join(ocr_pages).strip(), pages, True


# ---------------------------------------------------------------------------
# SECTION DETECTION
# ---------------------------------------------------------------------------

def normalize_heading(line: str) -> Optional[str]:
    raw = clean_line(line).lower().rstrip(":")
    raw = re.sub(r"[^a-z& ]", " ", raw)
    raw = re.sub(r"\s+", " ", raw).strip()

    for canonical, aliases in SECTION_ALIASES.items():
        if raw in [a.lower() for a in aliases]:
            return canonical
    return None


def detect_sections(text: str) -> Dict[str, List[str]]:
    sections = {key: [] for key in SECTION_ALIASES}
    current = None

    for raw in text.splitlines():
        line = clean_line(raw)
        if not line:
            continue

        heading = normalize_heading(line)
        if heading:
            current = heading
            continue

        if current:
            sections[current].append(line)

    return sections


# ---------------------------------------------------------------------------
# CONTACT / CANDIDATE INFO
# ---------------------------------------------------------------------------

def extract_candidate_info(text: str) -> Dict[str, Optional[str]]:
    email = re.search(r"\b[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}\b", text, re.I)
    phone = re.search(r"(?<!\d)(?:\+?\d[\d\s().-]{8,}\d)(?!\d)", text)

    name = None
    for line in text.splitlines()[:15]:
        s = strip_bullet(clean_line(line))
        if (
            s
            and not re.search(r"@", s)
            and not re.search(r"\b(?:resume|curriculum vitae|cv)\b", s, re.I)
            and len(s.split()) <= 6
            and re.fullmatch(r"[A-Za-z .'-]+", s)
        ):
            name = s
            break

    return {
        "name": name,
        "phone": phone.group(0).strip() if phone else None,
        "email": email.group(0).strip() if email else None,
    }


# ---------------------------------------------------------------------------
# JD PARSING
# ---------------------------------------------------------------------------

def extract_field(lines: List[str], labels: List[str]) -> Optional[str]:
    for i, line in enumerate(lines):
        clean = strip_bullet(line)
        for label in labels:
            m = re.match(rf"^{re.escape(label)}\s*:\s*(.*)$", clean, re.I)
            if m:
                value = m.group(1).strip()
                if value:
                    return value
                if i + 1 < len(lines):
                    nxt = strip_bullet(lines[i + 1])
                    if nxt:
                        return nxt
    return None


def extract_required_experience(text: str) -> Dict[str, Optional[float]]:
    patterns = [
        r"(\d+(?:\.\d+)?)\s*\+?\s*years?",
        r"minimum\s+(\d+(?:\.\d+)?)\s*years?",
    ]
    for p in patterns:
        m = re.search(p, text, re.I)
        if m:
            return {"min_years": float(m.group(1)), "max_years": None}
    return {"min_years": None, "max_years": None}


def split_skill_values(value: str) -> List[str]:
    parts = re.split(r",|;|\||\n", value)
    return [clean_text(x) for x in parts if clean_text(x)]


def extract_jd_skills(lines: List[str]) -> Tuple[List[str], List[str]]:
    technical = []
    soft = []

    in_required = False
    for i, raw in enumerate(lines):
        line = strip_bullet(raw)
        low = line.lower()

        if re.match(r"^required\s+skills?\s*:", low):
            in_required = True
            after = re.split(r":", line, 1)[1].strip()
            if after:
                values = split_skill_values(after)
                for value in values:
                    if any(re.search(p, value, re.I) for ps in SOFT_SKILL_ALIASES.values() for p in ps):
                        soft.append(value)
                    else:
                        technical.append(value)
            continue

        if in_required:
            # Stop when a new obvious field/section starts.
            if re.match(r"^(job role|job title|company|required experience|responsibilities|description)\s*:", low):
                in_required = False
                continue

            value = strip_bullet(line)
            if not value:
                continue

            values = split_skill_values(value)
            for value in values:
                if re.search(r"communication\s+skills?|leadership|teamwork", value, re.I):
                    soft.append(value)
                elif any(contains_skill(value, skill) for skill in SKILL_ALIASES):
                    technical.append(value)

    # Canonicalize recognized skills.
    def canonicalize(values: List[str]) -> List[str]:
        result = []
        for value in values:
            for canonical in SKILL_ALIASES:
                if contains_skill(value, canonical):
                    if canonical not in result:
                        result.append(canonical)
                    break
            else:
                if value not in result:
                    result.append(value)
        return result

    return canonicalize(technical), canonicalize(soft)


def parse_jd(jd_text: str) -> Dict:
    lines = [clean_line(x) for x in jd_text.splitlines() if clean_line(x)]

    company = extract_field(lines, ["Company"])
    role = extract_field(lines, ["Job Role", "Job Title"])
    required_exp = extract_required_experience(jd_text)
    skills, soft = extract_jd_skills(lines)

    return {
        "company": company,
        "job_title": role,
        "required_experience": required_exp,
        "required_skills": skills,
        "soft_skills": soft,
        "raw_text": jd_text.strip(),
    }


# ---------------------------------------------------------------------------
# CANDIDATE SKILL EXTRACTION + EVIDENCE
# ---------------------------------------------------------------------------

def extract_candidate_skills(
    sections: Dict[str, List[str]]
) -> Tuple[List[str], Dict[str, List[Dict]]]:

    evidence: Dict[str, List[Dict]] = {skill: [] for skill in SKILL_ALIASES}

    source_types = {
        "Skills": "DIRECT_SKILL",
        "Projects": "PROJECT_EVIDENCE",
        "Internship & Experience": "EXPERIENCE_EVIDENCE",
        "Professional Summary": "SUMMARY_EVIDENCE",
        "Certifications": "CERTIFICATION_EVIDENCE",
    }

    for source, lines in sections.items():
        if source not in source_types:
            continue

        text = remove_academic_abbreviations("\n".join(lines))

        for skill, patterns in SKILL_ALIASES.items():
            for pattern in patterns:
                for match in re.finditer(pattern, text, re.I):
                    evidence[skill].append({
                        "source": source,
                        "evidence_type": source_types[source],
                        "matched_text": match.group(0),
                        "context": evidence_context(text, match.start(), match.end()),
                    })
                    break

    candidate_skills = [
        skill for skill, items in evidence.items()
        if items
    ]

    # C should only be accepted when it has an actual standalone C evidence
    # occurrence outside academic abbreviations.
    if "C" in candidate_skills:
        candidate_skills = [
            s for s in candidate_skills
            if not (s == "C" and not any(
                e["source"] == "Skills" for e in evidence["C"]
            ))
        ]
        if not candidate_skills or "C" not in candidate_skills:
            evidence["C"] = []

    return candidate_skills, evidence


def compare_skills(
    required_skills: List[str],
    candidate_evidence: Dict[str, List[Dict]]
) -> Dict:
    matched = []
    missing = []
    details = []

    for required in required_skills:
        canonical = normalize_skill(required)
        matching_key = next(
            (k for k in candidate_evidence if normalize_skill(k) == canonical),
            required
        )
        items = candidate_evidence.get(matching_key, [])

        if items:
            matched.append(required)
            details.append({
                "skill": required,
                "status": "PRESENT",
                "evidence_types": sorted(set(e["evidence_type"] for e in items)),
                "evidence": items,
            })
        else:
            missing.append(required)
            details.append({
                "skill": required,
                "status": "MISSING",
                "evidence_types": [],
                "evidence": [],
            })

    score = round((len(matched) / len(required_skills)) * 100, 2) if required_skills else 0.0

    return {
        "matched_required_skills": matched,
        "missing_required_skills": missing,
        "required_skill_score": score,
        "match_details": details,
    }


# ---------------------------------------------------------------------------
# SOFT SKILLS
# ---------------------------------------------------------------------------

def soft_skill_evidence(
    sections: Dict[str, List[str]],
    required_soft_skills: List[str]
) -> Dict:

    results = {}

    searchable = []
    for section in ["Internship & Experience", "Professional Summary", "Projects", "Certifications"]:
        searchable.append((section, "\n".join(sections.get(section, []))))

    for required in required_soft_skills:
        key = next(
            (k for k in SOFT_SKILL_ALIASES if normalize_skill(k) == normalize_skill(required)),
            required
        )

        found = []
        for source, text in searchable:
            for pattern in SOFT_SKILL_ALIASES.get(key, []):
                for match in re.finditer(pattern, text, re.I):
                    # Do not count the exact JD phrase copied into the resume as
                    # strong proof unless the surrounding context is meaningful.
                    ctx = evidence_context(text, match.start(), match.end(), 120)
                    strength = "STRONG" if re.search(
                        r"\b(communicat|present|coordinat|collaborat|lead|manage|team|work(?:ed|ing)? with)\w*",
                        ctx, re.I
                    ) else "SUPPORTING"
                    found.append({
                        "source": source,
                        "strength": strength,
                        "matched_text": match.group(0),
                        "context": ctx,
                    })
                    break

        # Deduplicate by source/context.
        unique = []
        seen = set()
        for item in found:
            marker = (item["source"], item["context"])
            if marker not in seen:
                seen.add(marker)
                unique.append(item)

        if not unique:
            status = "NOT_FOUND"
            score = 0.0
        elif any(x["strength"] == "STRONG" for x in unique):
            status = "PRESENT"
            score = 100.0
        else:
            status = "SUPPORTING"
            score = 60.0

        results[required] = {
            "score": score,
            "status": status,
            "evidence": unique[:5],
        }

    overall = (
        round(sum(x["score"] for x in results.values()) / len(results), 2)
        if results else 0.0
    )

    return {
        "required_soft_skills": required_soft_skills,
        "results": results,
        "score": overall,
    }


# ---------------------------------------------------------------------------
# EXPERIENCE PARSING
# ---------------------------------------------------------------------------

def looks_like_role_heading(line: str, next_line: str = "") -> bool:
    raw = clean_line(line)
    stripped = strip_bullet(raw)
    low = stripped.lower()

    if not stripped or is_date_line(stripped):
        return False

    # Body sentences should not become role headings.
    if re.search(r"[.!?]$", stripped):
        return False

    # Strong structural signal: bullet role heading.
    if raw.startswith(("•", "●", "▪", "◦", "-", "*")):
        if any(low.startswith(v + " ") for v in ACTION_VERBS):
            return False
        return any(k in low for k in ROLE_KEYWORDS | NON_TECH_ROLE_KEYWORDS) or (
            len(stripped.split()) <= 12
        )

    # Non-bullet heading: only accept if followed by a date.
    if is_date_line(next_line):
        return any(k in low for k in ROLE_KEYWORDS | NON_TECH_ROLE_KEYWORDS)

    return False


def classify_role(role_title: str, body: str) -> str:
    text = f"{role_title} {body}".lower()

    if any(k in text for k in TECH_ROLE_KEYWORDS):
        return "TECHNICAL"

    if any(k in text for k in NON_TECH_ROLE_KEYWORDS):
        return "NON_TECHNICAL"

    # Do not infer technicality from generic words such as "technical event".
    return "OTHER"


def parse_experience(lines: List[str]) -> List[ExperienceRole]:
    roles: List[ExperienceRole] = []
    i = 0

    while i < len(lines):
        line = clean_line(lines[i])
        next_line = clean_line(lines[i + 1]) if i + 1 < len(lines) else ""

        if not looks_like_role_heading(line, next_line):
            i += 1
            continue

        title = strip_bullet(line)
        date_text = "Not specified"
        start = end = None
        certain = False
        current = False

        if i + 1 < len(lines) and is_date_line(next_line):
            date_text = next_line
            start, end, certain = parse_date_range(date_text)
            current = bool(re.search(r"\b(?:present|current)\b", date_text, re.I))
            body_start = i + 2
        else:
            body_start = i + 1

        body_lines = []
        j = body_start
        while j < len(lines):
            candidate = clean_line(lines[j])
            candidate_next = clean_line(lines[j + 1]) if j + 1 < len(lines) else ""

            if looks_like_role_heading(candidate, candidate_next):
                break

            # A date line directly after a role is consumed above; dates inside
            # a body are retained as body text.
            body_lines.append(candidate)
            j += 1

        body = clean_text(" ".join(body_lines))
        role_type = classify_role(title, body)

        roles.append(ExperienceRole(
            role_title=title,
            date_text=date_text,
            start_date=start,
            end_date=end,
            is_current=current,
            duration_certain=certain,
            role_type=role_type,
            body=body,
        ))

        i = j

    return roles


def months_between(start: str, end: str) -> int:
    sy, sm = int(start[:4]), int(start[5:7])
    ey, em = int(end[:4]), int(end[5:7])
    return max(0, (ey - sy) * 12 + (em - sm) + 1)


def calculate_technical_experience(roles: List[ExperienceRole]) -> Dict:
    technical = [r for r in roles if r.role_type == "TECHNICAL"]

    certain_months = 0
    uncertain = False

    for role in technical:
        if role.start_date and role.end_date and role.duration_certain:
            certain_months += months_between(role.start_date, role.end_date)
        else:
            uncertain = True

    if not technical:
        return {
            "status": "NOT_SPECIFIED",
            "total_months": 0,
            "total_years": 0.0,
            "technical_roles": [],
        }

    if uncertain and certain_months == 0:
        status = "NOT_SPECIFIED"
        total_months = None
        total_years = None
    elif uncertain:
        status = "PARTIAL"
        total_months = certain_months
        total_years = round(certain_months / 12, 2)
    else:
        status = "DETERMINED"
        total_months = certain_months
        total_years = round(certain_months / 12, 2)

    return {
        "status": status,
        "total_months": total_months,
        "total_years": total_years,
        "technical_roles": [asdict(r) for r in technical],
    }


def experience_evidence(
    roles: List[ExperienceRole],
    job_title: Optional[str],
) -> Tuple[Dict, float]:

    technical_roles = [r for r in roles if r.role_type == "TECHNICAL"]
    text = " ".join(
        f"{r.role_title} {r.body}" for r in technical_roles
    )

    patterns = {
        "full stack development": [
            r"\bfull[\s-]?stack\b",
        ],
        "web application development": [
            r"\bweb\s+(?:application|platform)\b",
            r"\bweb\s+development\b",
        ],
        "frontend development": [
            r"\bfrontend\b", r"\bfront[-\s]?end\b",
        ],
        "backend development": [
            r"\bbackend\b", r"\bback[-\s]?end\b",
            r"\bserver[-\s]?side\b",
        ],
        "REST API development": [
            r"\bdevelop(?:ed|ing)?\s+(?:a\s+)?rest\s+apis?\b",
            r"\bimplement(?:ed|ing)?\s+(?:a\s+)?rest\s+apis?\b",
            r"\brestful\s+apis?\b",
        ],
        "database development": [
            r"\bdatabase\b",
            r"\bsql\b",
            r"\bmongodb\b",
            r"\bmysql\b",
        ],
    }

    evidence = {}
    numeric_values = []

    for label, pats in patterns.items():
        hits = []
        for p in pats:
            for m in re.finditer(p, text, re.I):
                hits.append(evidence_context(text, m.start(), m.end(), 160))

        hits = list(dict.fromkeys(hits))
        if hits:
            # Explicit technical wording = strong; generic database mention = supporting.
            strong = any(
                re.search(r"\b(develop|developed|implement|implemented|build|built|create|created)\w*", h, re.I)
                for h in hits
            )
            strength = "STRONG" if strong else "SUPPORTING"
            value = 1.0 if strong else 0.5
        else:
            strength = "NONE"
            value = 0.0

        evidence[label] = {
            "evidence": hits[:3],
            "strength": strength,
            "numeric_score": value,
        }
        numeric_values.append(value)

    # A technical internship is real technical experience even if its body
    # does not prove a specific backend/API/database technology.
    baseline = 0.35 if technical_roles else 0.0
    specific = (sum(numeric_values) / len(numeric_values)) if numeric_values else 0.0

    # Keep the baseline and evidence separate, so we never inflate evidence.
    score = round(max(baseline, specific * 0.65 + baseline * 0.35) * 100, 2)

    return evidence, score


# ---------------------------------------------------------------------------
# PROJECT PARSING
# ---------------------------------------------------------------------------

def looks_like_project_title(line: str) -> bool:
    raw = clean_line(line)
    stripped = strip_bullet(raw)

    if not stripped or is_year_only(stripped):
        return False
    if re.match(r"^tech(?:nologies)?\s*:", stripped, re.I):
        return False

    # Current resume format uses bullet headings. A non-bullet line is only
    # considered a title if it looks unmistakably like a heading.
    if raw.startswith(("•", "●", "▪", "◦", "-", "*")):
        if re.match(r"^(built|developed|implemented|created|designed|analyzed|analysed|using|used)\b", stripped, re.I):
            return False
        return len(stripped) <= 160

    return False


def extract_inline_title_and_description(value: str) -> Tuple[str, str]:
    # Project titles can themselves contain punctuation. Split only at the
    # first colon followed by an action verb.
    m = re.match(
        r"^(.*?):\s*((?:built|developed|implemented|created|designed|analyzed|analysed|refactored|engineered|used|using)\b.*)$",
        value,
        re.I,
    )
    if m:
        return clean_text(m.group(1)), clean_text(m.group(2))

    return clean_text(value), ""


def parse_tech_line(line: str) -> List[str]:
    value = re.sub(r"^\s*Tech(?:nologies)?\s*:\s*", "", line, flags=re.I)
    return [
        clean_text(x).rstrip(".")
        for x in re.split(r",|;|\|", value)
        if clean_text(x)
    ]


def parse_projects(lines: List[str]) -> List[Project]:
    projects: List[Project] = []
    current: Optional[Project] = None

    for raw in lines:
        line = clean_line(raw)
        if not line:
            continue

        if re.match(r"^Tech(?:nologies)?\s*:", line, re.I):
            if current:
                current.technologies.extend(parse_tech_line(line))
                current.technologies = list(dict.fromkeys(current.technologies))
            continue

        if is_year_only(line):
            if current and not current.year:
                current.year = line
            continue

        if looks_like_project_title(line):
            title_value = strip_bullet(line)
            title, desc = extract_inline_title_and_description(title_value)

            current = Project(
                title=title,
                technologies=[],
                year=None,
                description=desc,
            )
            projects.append(current)
            continue

        if current:
            current.description = clean_text(
                f"{current.description} {line}"
            ).strip()

    return projects


# ---------------------------------------------------------------------------
# PROJECT MATCHING
# ---------------------------------------------------------------------------

def project_skill_overlap(project: Project, required_skills: List[str]) -> Tuple[List[str], List[str], float]:
    matched = []
    missing = []

    combined = " ".join([
        project.title,
        project.description,
        " ".join(project.technologies),
    ])

    for skill in required_skills:
        if contains_skill(combined, skill):
            matched.append(skill)
        else:
            missing.append(skill)

    score = round((len(matched) / len(required_skills)) * 100, 2) if required_skills else 0.0
    return matched, missing, score


def implementation_score(project: Project) -> float:
    text = f"{project.title} {project.description}".lower()
    action = bool(re.search(
        r"\b(built|developed|implemented|created|designed|deployed|refactored|integrated|analyzed|analysed)\b",
        text,
    ))
    measurable = bool(re.search(r"\b\d+(?:\.\d+)?\s*%", text))
    if action and measurable:
        return 100.0
    if action:
        return 75.0
    return 25.0


def role_relevance_score(
    project: Project,
    job_title: Optional[str],
    required_skills: List[str]
) -> float:

    role = (job_title or "").lower()
    text = f"{project.title} {project.description} {' '.join(project.technologies)}".lower()

    if not role:
        return 0.0

    score = 0.0

    # General software-development relevance.
    if any(k in role for k in ["software", "developer", "engineer", "application"]):
        if any(k in text for k in [
            "system", "software", "application", "platform", "code",
            "development", "developed", "built"
        ]):
            score += 40.0

    if "web" in role and re.search(r"\bweb\b", text):
        score += 20.0

    if "backend" in role and re.search(r"\bbackend\b|\bback-end\b", text):
        score += 25.0

    if "frontend" in role and re.search(r"\bfrontend\b|\bfront-end\b", text):
        score += 25.0

    # Direct JD overlap adds relevance but cannot by itself make every project 100%.
    matched, _, overlap = project_skill_overlap(project, required_skills)
    if required_skills:
        score += min(30.0, overlap * 0.30)

    # Cap: relevance is evidence-based, not a binary "all projects relevant".
    return round(min(score, 100.0), 2)


def project_generic_evidence(project: Project) -> Tuple[Dict, float]:
    text = f"{project.title} {project.description} {' '.join(project.technologies)}"

    categories = {
        "full stack development": [r"\bfull[\s-]?stack\b"],
        "web application development": [
            r"\bweb\s+(?:application|platform)\b",
            r"\bweb\s+development\b",
        ],
        "frontend development": [
            r"\bfrontend\b", r"\bfront[-\s]?end\b", r"\breact\b"
        ],
        "backend development": [
            r"\bbackend\b", r"\bback[-\s]?end\b",
            r"\bflask\b", r"\bfastapi\b", r"\bdjango\b",
            r"\bnode(?:\.js|js)?\b", r"\bexpress(?:\.js|js)?\b",
        ],
        "REST API development": [
            r"\brest\s+apis?\b", r"\brestful\s+apis?\b"
        ],
        "database development": [
            r"\bdatabase\b", r"\bmongodb\b", r"\bmysql\b", r"\bsql\b",
            r"\bfirebase\b", r"\bconvex\b",
        ],
    }

    result = {}
    values = []

    for label, pats in categories.items():
        hits = []
        for p in pats:
            for m in re.finditer(p, text, re.I):
                hits.append(evidence_context(text, m.start(), m.end(), 150))

        hits = list(dict.fromkeys(hits))
        if not hits:
            strength, value = "NONE", 0.0
        elif any(re.search(r"\b(develop|developed|build|built|implement|implemented)\w*", h, re.I) for h in hits):
            strength, value = "STRONG", 1.0
        else:
            strength, value = "PARTIAL", 0.5

        result[label] = {
            "evidence": hits[:3],
            "strength": strength,
            "numeric_score": value,
        }
        values.append(value)

    score = round((sum(values) / len(values)) * 100, 2) if values else 0.0
    return result, score


def evaluate_project(
    project: Project,
    job_title: Optional[str],
    required_skills: List[str]
) -> Dict:

    evidence, generic_score = project_generic_evidence(project)
    matched, missing, overlap_score = project_skill_overlap(project, required_skills)
    relevance = role_relevance_score(project, job_title, required_skills)
    implementation = implementation_score(project)

    # Balanced project score.
    score = round(
        generic_score * 0.20
        + overlap_score * 0.35
        + relevance * 0.30
        + implementation * 0.15,
        2,
    )

    return {
        "title": project.title,
        "technologies": project.technologies,
        "year": project.year,
        "description": project.description,
        "evidence": evidence,
        "generic_project_evidence_score": generic_score,
        "jd_matched_skills": matched,
        "jd_missing_skills": missing,
        "jd_skill_overlap_score": overlap_score,
        "role_relevance_score": relevance,
        "implementation_evidence_score": implementation,
        "project_match_score": score,
    }


# ---------------------------------------------------------------------------
# OVERALL EVALUATION
# ---------------------------------------------------------------------------

def evaluate_experience_eligibility(
    experience_info: Dict,
    required_experience: Dict
) -> str:

    minimum = required_experience.get("min_years")
    actual = experience_info.get("total_years")

    if minimum is None:
        return "NOT_SPECIFIED"

    if actual is None:
        return "UNKNOWN"

    return "ELIGIBLE" if actual >= minimum else "NOT_ELIGIBLE"


def build_final_result(
    resume_text: str,
    jd_text: str,
    sections: Dict[str, List[str]],
    candidate: Dict,
    jd: Dict,
) -> Dict:

    candidate_skills, skill_evidence = extract_candidate_skills(sections)
    skill_result = compare_skills(jd["required_skills"], skill_evidence)

    soft_result = soft_skill_evidence(sections, jd["soft_skills"])

    roles = parse_experience(sections.get("Internship & Experience", []))
    exp_duration = calculate_technical_experience(roles)
    eligibility = evaluate_experience_eligibility(
        exp_duration, jd["required_experience"]
    )
    exp_evidence, exp_score = experience_evidence(roles, jd["job_title"])

    projects = parse_projects(sections.get("Projects", []))
    project_results = [
        evaluate_project(p, jd["job_title"], jd["required_skills"])
        for p in projects
    ]
    project_score = round(
        sum(p["project_match_score"] for p in project_results) / len(project_results),
        2
    ) if project_results else 0.0

    # Soft skills are reported separately and are not double-counted.
    overall = round(
        skill_result["required_skill_score"] * 0.50
        + exp_score * 0.20
        + project_score * 0.30,
        2,
    )

    status = (
        "HIGH MATCH" if overall >= 75
        else "MEDIUM MATCH" if overall >= 50
        else "LOW MATCH"
    )

    return {
        "candidate": candidate,
        "job": jd,
        "candidate_fit": {
            "overall_match_score": overall,
            "overall_status": status,
        },
        "soft_skill_analysis": soft_result,
        "skills": {
            "candidate_skills": candidate_skills,
            **skill_result,
            "skill_evidence": skill_evidence,
        },
        "experience": {
            "experience_duration": exp_duration,
            "experience_eligibility": eligibility,
            "experience_evidence_score": exp_score,
            "technical_experience_text": " ".join(
                f"{r.role_title} {r.body}" for r in roles if r.role_type == "TECHNICAL"
            ),
            "evidence": exp_evidence,
            "roles": [asdict(r) for r in roles],
        },
        "projects": {
            "total_projects": len(project_results),
            "projects": project_results,
            "project_match_score": project_score,
        },
        "score_breakdown": {
            "skills": {
                "score": skill_result["required_skill_score"],
                "weight": 0.50,
                "contribution": round(skill_result["required_skill_score"] * 0.50, 2),
            },
            "experience": {
                "score": exp_score,
                "weight": 0.20,
                "contribution": round(exp_score * 0.20, 2),
            },
            "projects": {
                "score": project_score,
                "weight": 0.30,
                "contribution": round(project_score * 0.30, 2),
            },
            "soft_skills": {
                "score": soft_result["score"],
                "weight": 0.0,
                "contribution": 0.0,
                "note": "Reported separately to avoid double-counting with technical fit.",
            },
        },
        "resume": {
            "sections": sections,
        },
    }


# ---------------------------------------------------------------------------
# REPORTING
# ---------------------------------------------------------------------------

def print_report(result: Dict) -> None:
    print("=" * 80)
    print("RESUME–JD MATCHING SYSTEM — V9 ACCURACY FIXED")
    print("=" * 80)

    candidate = result["candidate"]
    job = result["job"]

    print("\nCANDIDATE INFORMATION")
    print("-" * 80)
    print(f"Name  : {candidate.get('name') or 'Not detected'}")
    print(f"Phone : {candidate.get('phone') or 'Not detected'}")
    print(f"Email : {candidate.get('email') or 'Not detected'}")

    print("\nJOB INFORMATION")
    print("-" * 80)
    print(f"Company: {job.get('company') or 'Not specified'}")
    print(f"Job Title: {job.get('job_title') or 'Not specified'}")
    print(f"Required Experience: {job['required_experience']}")

    print("\nRequired Technical Skills:")
    for x in job["required_skills"]:
        print(f"  • {x}")

    print("\nCANDIDATE SKILLS")
    print("-" * 80)
    for x in result["skills"]["candidate_skills"]:
        print(f"  • {x}")

    print("\nMatched Required Skills:")
    for x in result["skills"]["matched_required_skills"]:
        print(f"  ✓ {x}")

    print("\nMissing Required Skills:")
    for x in result["skills"]["missing_required_skills"]:
        print(f"  ✗ {x}")

    print(f"\nTechnical Skill Match Score: {result['skills']['required_skill_score']:.2f}%")

    print("\nSOFT SKILL ANALYSIS")
    print("-" * 80)
    if not result["soft_skill_analysis"]["required_soft_skills"]:
        print("  No soft skills specified in JD.")
    else:
        for skill, data in result["soft_skill_analysis"]["results"].items():
            print(f"  {skill}: {data['status']} ({data['score']:.2f}%)")
            for e in data["evidence"][:2]:
                print(f"      [{e['strength']}] {e['context']}")

    print("\nEXPERIENCE ANALYSIS")
    print("-" * 80)
    roles = result["experience"]["roles"]
    if not roles:
        print("  No reliable experience roles detected.")
    for idx, role in enumerate(roles, 1):
        print(f"  Role {idx}: {role['role_title']}")
        print(f"    Company: {role['company']}")
        print(f"    Dates: {role['date_text']}")
        print(f"    Classification: {role['role_type']}")

    duration = result["experience"]["experience_duration"]
    print(f"\nTechnical Experience Status: {duration['status']}")
    print(f"Technical Experience Duration: {duration['total_years']}")
    print(f"Experience Eligibility: {result['experience']['experience_eligibility']}")
    print(f"Experience Evidence Score: {result['experience']['experience_evidence_score']:.2f}%")

    print("\nPROJECT ANALYSIS")
    print("-" * 80)
    print(f"Total Projects Detected: {result['projects']['total_projects']}")
    for idx, project in enumerate(result["projects"]["projects"], 1):
        print(f"\n  Project {idx}: {project['title']}")
        print(f"    Technologies: {', '.join(project['technologies']) or 'None'}")
        print(f"    Description: {project['description']}")
        print(f"    JD Skill Overlap: {project['jd_skill_overlap_score']:.2f}%")
        print(f"    Role Relevance: {project['role_relevance_score']:.2f}%")
        print(f"    Implementation Evidence: {project['implementation_evidence_score']:.2f}%")
        print(f"    Project Match Score: {project['project_match_score']:.2f}%")

    print(f"\nOverall Project Score: {result['projects']['project_match_score']:.2f}%")

    print("\nFINAL RESULT")
    print("-" * 80)
    print(f"Overall Match Score: {result['candidate_fit']['overall_match_score']:.2f}%")
    print(f"Overall Status: {result['candidate_fit']['overall_status']}")

    print("\n" + "=" * 80)
    print("STRUCTURED JSON OUTPUT")
    print("=" * 80)
    print(json.dumps(result, indent=4, ensure_ascii=False))


# ---------------------------------------------------------------------------
# FILE LOADING / MAIN
# ---------------------------------------------------------------------------

def load_jd_file(path: Path) -> str:
    return path.read_text(encoding="utf-8")


def run_evaluation(resume_path, jd_path, output_path=None):
    """
    Run the complete Resume-JD evaluation.

    This function is the recommended way to run the system
    from Jupyter Notebook.
    """

    resume_path = Path(resume_path)
    jd_path = Path(jd_path)

    if not resume_path.exists():
        raise FileNotFoundError(
            f"Resume not found: {resume_path}"
        )

    if not jd_path.exists():
        raise FileNotFoundError(
            f"JD not found: {jd_path}"
        )

    # ---------------------------------------------------------
    # STEP 1: Extract resume text
    # ---------------------------------------------------------
    resume_text, pages, ocr_used = extract_pdf_text(resume_path)

    # ---------------------------------------------------------
    # STEP 2: Read JD
    # ---------------------------------------------------------
    jd_text = load_jd_file(jd_path)

    # ---------------------------------------------------------
    # STEP 3: Detect resume sections
    # ---------------------------------------------------------
    sections = detect_sections(resume_text)

    # ---------------------------------------------------------
    # STEP 4: Extract candidate information
    # ---------------------------------------------------------
    candidate = extract_candidate_info(resume_text)

    # ---------------------------------------------------------
    # STEP 5: Parse JD
    # ---------------------------------------------------------
    jd = parse_jd(jd_text)

    # ---------------------------------------------------------
    # STEP 6: Evaluate candidate
    # ---------------------------------------------------------
    result = build_final_result(
        resume_text,
        jd_text,
        sections,
        candidate,
        jd,
    )

    # ---------------------------------------------------------
    # BASIC INFORMATION
    # ---------------------------------------------------------
    print("=" * 80)
    print("RESUME-JD MATCHING SYSTEM")
    print("V9 - ACCURACY FIXED")
    print("=" * 80)

    print()
    print(f"Resume: {resume_path}")
    print(f"JD: {jd_path}")
    print(f"Resume pages: {pages}")
    print(f"OCR used: {ocr_used}")
    print(f"Extracted characters: {len(resume_text)}")

    # ---------------------------------------------------------
    # RESUME SECTIONS
    # ---------------------------------------------------------
    print()
    print("=" * 80)
    print("DETECTED RESUME SECTIONS")
    print("=" * 80)

    for section_name, section_lines in sections.items():
        if section_lines:
            print(
                f"{section_name}: "
                f"{len(section_lines)} lines"
            )

    # ---------------------------------------------------------
    # FINAL REPORT
    # ---------------------------------------------------------
    print_report(result)

    # ---------------------------------------------------------
    # SAVE JSON
    # ---------------------------------------------------------
    if output_path is None:
        output_path = resume_path.parent / "candidate_evaluation_result.json"
    else:
        output_path = Path(output_path)

    output_path.write_text(
        json.dumps(
            result,
            indent=4,
            ensure_ascii=False
        ),
        encoding="utf-8"
    )

    print()
    print(f"JSON result saved to: {output_path}")

    return result


def main():
    """
    Command-line entry point.

    Example:

    python InterviewQuestionGenerator_v9_accuracy_fixed.py ^
        --resume ..\\data\\Sumit_JoshSoftware_resume.pdf ^
        --jd ..\\data\\sample_jd.txt
    """

    parser = argparse.ArgumentParser(
        description=(
            "Interview Expert - "
            "Resume/JD Matching System"
        )
    )

    parser.add_argument(
        "--resume",
        required=True,
        help="Path to resume PDF"
    )

    parser.add_argument(
        "--jd",
        required=True,
        help="Path to JD text file"
    )

    parser.add_argument(
        "--output",
        default=None,
        help="Optional JSON output path"
    )

    args = parser.parse_args()

    run_evaluation(
        resume_path=args.resume,
        jd_path=args.jd,
        output_path=args.output
    )


# Only run command-line arguments when this file is
# executed directly.
#
# When imported into Jupyter Notebook, main() will not run.
if __name__ == "__main__":
    main()
```
